```
import math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
from scipy.stats import norm
from collections import defaultdict

# --- SAX Algorithm Class ---
class SAXAlgorithm:
    @staticmethod
    def z_normalize(series):
        if len(series) == 0:
            return np.array([])
        mean = np.mean(series)
        std = np.std(series)
        if std == 0:
            return np.zeros_like(series)
        return (series - mean) / std

    @staticmethod
    def paa(series, num_segments):
        if num_segments <= 0:
            raise ValueError("num_segments must be positive")
        n = len(series)
        if n == 0:
            return np.full(num_segments, np.nan)
        segments = np.array_split(series, num_segments)
        return np.array([np.mean(seg) for seg in segments])

    @staticmethod
    def get_breakpoints(alphabet_size):
        if not isinstance(alphabet_size, int) or alphabet_size < 1:
            raise ValueError("alphabet_size must be a positive integer")
        if alphabet_size <= 1:
            return np.array([])
        return norm.ppf(np.linspace(1/alphabet_size, 1 - 1/alphabet_size, alphabet_size - 1))

    @staticmethod
    def series_to_sax(series, word_length, alphabet_size, segment_idx=None):
        if word_length <= 0 or alphabet_size <= 0:
            raise ValueError("word_length and alphabet_size must be positive")
        if alphabet_size > 26:
            raise ValueError("alphabet_size must not exceed 26 for lowercase letters")
        if not isinstance(series, np.ndarray):
            series_np = np.array(series, dtype=np.float64)
        else:
            series_np = series.astype(np.float64)
        if len(series_np) == 0:
            return "!" * word_length
        paa_values = SAXAlgorithm.paa(series_np, word_length)
        if np.isnan(paa_values).any():
            if np.all(np.isnan(paa_values)):
                return "?" * word_length
            paa_values = np.nan_to_num(paa_values, nan=0.0)
        breakpoints = SAXAlgorithm.get_breakpoints(alphabet_size)
        if segment_idx is not None:
            print(f"Segment {segment_idx} PAA values: {paa_values}")
            print(f"Breakpoints: {breakpoints}")
        alphabet = [chr(ord('a') + i) for i in range(alphabet_size)]
        sax_list = []
        for val in paa_values:
            symbol = alphabet[np.searchsorted(breakpoints, val)]
            sax_list.append(symbol)
        sax_str = "".join(sax_list)
        if segment_idx is not None:
            print(f"Segment {segment_idx} SAX string: {sax_str}")
        return sax_str

    @staticmethod
    def levenshtein_distance(s1, s2):
        if len(s1) < len(s2):
            return SAXAlgorithm.levenshtein_distance(s2, s1)
        if len(s2) == 0:
            return len(s1)
        prev_row = range(len(s2) + 1)
        for i, c1 in enumerate(s1):
            curr_row = [i + 1]
            for j, c2 in enumerate(s2):
                subs = prev_row[j] + (c1 != c2)
                curr_row.append(min(prev_row[j+1]+1, curr_row[j]+1, subs))
            prev_row = curr_row
        return prev_row[-1]

    @staticmethod
    def count_symbol_changes(sax_str):
        if not sax_str or len(sax_str) < 2:
            return 0
        changes = 0
        for i in range(len(sax_str) - 1):
            if sax_str[i] != sax_str[i+1]:
                changes += 1
        return changes

    @staticmethod
    def get_sax_symbol_range(sax_str):
        if not sax_str:
            return 0
        valid_chars = [s for s in sax_str if s.islower()]
        if not valid_chars:
            return 0
        ordinal_values = [ord(s) - ord('a') for s in valid_chars]
        return max(ordinal_values) - min(ordinal_values)

    @staticmethod
    def get_sax_value_volatility(sax_str):
        if not sax_str or len(sax_str) < 2:
            return 0
        valid_chars = [s for s in sax_str if s.islower()]
        if len(valid_chars) < 2:
            return 0
        ordinal_values = [ord(s) - ord('a') for s in valid_chars]
        volatility = sum(abs(ordinal_values[i+1] - ordinal_values[i]) for i in range(len(ordinal_values)-1))
        return volatility

# --- Normalization Function Class ---
class NormalizeFunction:
    @staticmethod
    def tanh1_np(x_np):
        if not isinstance(x_np, np.ndarray):
            x_np = np.array(x_np, dtype=np.float64)
        x_scaled = x_np * 1e-5
        v_clip = np.clip(x_scaled, -20.0, 20.0)
        out = np.tanh(v_clip)
        if len(out) > 1:
            out[1:] *= 2
        return out * 10000

    @staticmethod
    def tanh2_np(x_np):
        if not isinstance(x_np, np.ndarray):
            x_np = np.array(x_np, dtype=np.float64)
        x_scaled = x_np * 1e-5
        v_clip = np.clip(x_scaled, -20.0, 20.0)
        out = np.tanh(v_clip)
        return out * 100

    @staticmethod
    def tanh3_np(x_np):
        if not isinstance(x_np, np.ndarray):
            x_np = np.array(x_np, dtype=np.float64)
        x_scaled = x_np * 1e-5
        v_clip = np.clip(x_scaled, -20.0, 20.0)
        out = np.tanh(v_clip)
        return out * 1000

# --- Function to Remove Non-Overlapping Segments ---
def remove_non_overlapping_segment(data_dict, segment_to_remove=8, sax_column='SAX'):
    segment_sax_patterns = {}
    for dataset_name, segments in data_dict.items():
        for segment in segments:
            if segment.get('Segment') == f"Segment {segment_to_remove}":
                segment_sax_patterns[dataset_name] = segment[sax_column]
                break
        else:
            segment_sax_patterns[dataset_name] = None

    sax_patterns = set([pattern for pattern in segment_sax_patterns.values() if pattern is not None])
    if len(sax_patterns) == 0:
        print(f"No Segment {segment_to_remove} found in any dataset.")
        return data_dict
    elif len(sax_patterns) == 1:
        print(f"Segment {segment_to_remove} has identical SAX patterns across datasets. No removal needed.")
        return data_dict

    datasets_to_remove = []
    all_patterns = list(segment_sax_patterns.values())
    for dataset_name, pattern in segment_sax_patterns.items():
        if pattern is None:
            continue
        pattern_count = sum(1 for p in all_patterns if p == pattern)
        if pattern_count == 1:
            datasets_to_remove.append(dataset_name)

    filtered_data_dict = data_dict.copy()
    for dataset_name in datasets_to_remove:
        segments = filtered_data_dict[dataset_name]
        filtered_data_dict[dataset_name] = [
            segment for segment in segments if segment.get('Segment') != f"Segment {segment_to_remove}"
        ]
        print(f"Removed Segment {segment_to_remove} from {dataset_name} (non-overlapping SAX pattern).")

    return filtered_data_dict

# --- Functions to Compute SAX Statistics ---
def sax_to_numerical(sax_string):
    numerical_values = []
    for char in sax_string:
        if char.islower():
            value = ord(char) - ord('a')
            numerical_values.append(value)
    return numerical_values

def compute_stats(sax_string):
    values = sax_to_numerical(sax_string)
    if not values:
        return {'mean': np.nan, 'std': np.nan}
    mean = np.mean(values)
    std = np.std(values)
    return {'mean': mean, 'std': std}

# --- Function to Identify Outlier Segments ---
def identify_outlier_segments(sax_data):
    outliers = {}
    for dataset_name, segments in sax_data.items():
        metrics = np.array([
            [s['Volatility'], s['Mean'], s['Std']] for s in segments
        ])
        mean_metrics = np.mean(metrics, axis=0)
        std_metrics = np.std(metrics, axis=0)
        std_metrics[std_metrics == 0] = 1.0
        standardized_metrics = (metrics - mean_metrics) / std_metrics
        mean_standardized = np.mean(standardized_metrics, axis=0)
        scores = np.sqrt(np.sum((standardized_metrics - mean_standardized) ** 2, axis=1))
        outlier_idx = np.argmax(scores)
        outlier_segment = segments[outlier_idx]
        outlier_score = scores[outlier_idx]
        outliers[dataset_name] = {
            'Segment': outlier_segment['Segment'],
            'SAX': outlier_segment['SAX'],
            'Volatility': outlier_segment['Volatility'],
            'Mean': outlier_segment['Mean'],
            'Std': outlier_segment['Std'],
            'Outlier Score': outlier_score,
            'Scores': scores.tolist()
        }
    return outliers

# --- Function to Remove Segments with Low Std ---
def remove_low_std_segments(data_dict, std_threshold=1.0):
    segments_to_remove = set()
    for dataset_name, segments in data_dict.items():
        for segment in segments:
            if segment['Std'] < std_threshold:
                segments_to_remove.add(segment['Segment'])

    if not segments_to_remove:
        print("No segments with Std < {} found across datasets.".format(std_threshold))
        return data_dict

    filtered_data_dict = data_dict.copy()
    for dataset_name in filtered_data_dict:
        segments = filtered_data_dict[dataset_name]
        filtered_data_dict[dataset_name] = [
            segment for segment in segments if segment['Segment'] not in segments_to_remove
        ]
        for segment_name in segments_to_remove:
            print(f"Removed {segment_name} from {dataset_name} (Std < {std_threshold}).")

    return filtered_data_dict

# --- File Paths ---
data_dir = '/content/drive/MyDrive/watch_ring/250429'
right_ring_path = os.path.join(data_dir + '/04April_2011_Monday_heute-956/right_ring/RingSpO2_Id000001052_Dev516C_NoDev_1745891550572_P09_20250429_105230_HRIBI.csv')
left_ring_path = os.path.join(data_dir +'/04April_2011_Monday_heute-956/left_ring/RingSpO2_Id000001052_DevF952_NoDev_1745891555676_P09_20250429_105235_HRIBI.csv')
right_watch_imu = os.path.join(data_dir +'/04April_2011_Monday_heute-956/left_watch/fusion_20240416-120255.csv')
right_watch_ppg = os.path.join(data_dir +'/04April_2011_Monday_heute-956/left_watch/AFELogging_R945USQU1AXC2_6H0K_1.0.57_20240416-120255.csv')
left_watch_imu = os.path.join(data_dir +'/04April_2011_Monday_heute-956/right_watch/fusion_20240617-150340.csv')
left_watch_ppg = os.path.join(data_dir +'/04April_2011_Monday_heute-956/right_watch/AFELogging_R945USQU1AXC2_6HHY_1.0.57_20240617-150340.csv')

# --- Channel Definitions ---
right_ring_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'PPG_G1', 'PPG_G2', 'PPG_R1', 'PPG_R2', 'PPG', 'PPG_IR2']
left_ring_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'PPG_G1', 'PPG_G2', 'PPG_R1', 'PPG_R2', 'PPG', 'PPG_IR2']
right_watch_imu_channels = ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z', 'GYR_X(degree/sec)', 'GYR_Y', 'GYR_Z', 'Mag_X', 'Mag_Y', 'Mag_Z']
right_watch_ppg_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'adc5', 'adc7']
left_watch_imu_channels = ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z', 'GYR_X(degree/sec)', 'GYR_Y', 'GYR_Z', 'Mag_X', 'Mag_Y', 'Mag_Z']
left_watch_ppg_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'adc5', 'adc7']

# --- Reusable Functions ---
def extract_acc_segments(df, channels, window, std_q, min_dur, val_thresh, length_thresh):
    if not all(ch in df.columns for ch in channels):
        print(f"  Error: Not all required Acc channels ({channels}) found in DataFrame columns: {df.columns.tolist()}")
        return []
    try:
        print(f"  Acc data stats: {df[channels].describe()}")
        rs = df[channels].rolling(window=window, center=True).std().ffill().bfill().fillna(0)
        print(f"  Rolling std stats: {rs.describe()}")
        th = rs.quantile(std_q)
        print(f"  Std threshold (quantile {std_q}): {th}")
        calm_conditions = []
        for ch in channels:
            if pd.isna(th[ch]):
                print(f"    Warning: Could not compute threshold for channel '{ch}'. Skipping channel for calm mask.")
                calm_conditions.append(pd.Series([False] * len(df), index=df.index))
            else:
                calm_conditions.append(rs[ch] < th[ch])
        if not calm_conditions:
            print("    Error: Could not compute thresholds for any Acc channel.")
            return []
        calm_mask = np.logical_and.reduce(calm_conditions)
        print(f"  Calm mask: {calm_mask.sum()} true values out of {len(calm_mask)}")
        calm_mask = pd.Series(calm_mask, index=df.index)
        diff = calm_mask.astype(int).diff().fillna(0)
        starts = np.where(diff == 1)[0]
        ends = np.where(diff == -1)[0]
        if len(calm_mask) > 0:
            if calm_mask.iloc[0]:
                starts = np.insert(starts, 0, 0)
            if len(ends) < len(starts) and calm_mask.iloc[-1]:
                ends = np.append(ends, len(calm_mask) - 1)
        min_len = min(len(starts), len(ends))
        starts = starts[:min_len]
        ends = ends[:min_len]
        segs = []
        for s, e in zip(starts, ends):
            if s >= e:
                continue
            length = e - s + 1
            if length < min_dur:
                continue
            segment_data = df[channels].iloc[s : e + 1]
            max_abs = segment_data.fillna(0).abs().max().max()
            if max_abs <= val_thresh:
                print(f"  Segment ({s}, {e}) discarded: max abs value {max_abs:.2f} <= {val_thresh}")
                continue
            if length <= length_thresh:
                print(f"  Segment ({s}, {e}) discarded: length {length} <= {length_thresh}")
                continue
            segs.append((s, e))
        return segs
    except Exception as e:
        print(f"Error during segment extraction: {e}")
        return []

# --- Modified Main Processing Function with SAX ---
def process_and_plot(df, label, params, acc_channels, all_channels, sampling_rate, sax_data_dict):
    print(f"\n===== Processing: {label} (Sampling Rate: {sampling_rate}Hz) =====")
    if df is None or df.empty:
        print("Input DataFrame is empty or None. Skipping.")
        return

    print(f"Input data shape: {df.shape}")
    print(f"Using parameters: {params}")
    print(f"Using Acc channels: {acc_channels}")
    print(f"Using all channels: {all_channels}")

    acc_segs = extract_acc_segments(
        df, acc_channels,
        window=params['WINDOW'],
        std_q=params['STD_QUANTILE'],
        min_dur=params['MIN_DURATION'],
        val_thresh=params['VAL_THRESH'],
        length_thresh=params['LENGTH_THRESH']
    )

    if not acc_segs:
        print(f"No valid segments found for {label}. Skipping SAX and plots.")
        return

    sax_channel = acc_channels[0]
    if sax_channel not in df.columns:
        print(f"  Error: SAX channel '{sax_channel}' not found. Skipping SAX analysis.")
        return

    normalized_acc_x = SAXAlgorithm.z_normalize(df[sax_channel].to_numpy())
    df[sax_channel + '_normalized'] = normalized_acc_x
    print(f"Normalized {sax_channel} stats: mean={np.mean(normalized_acc_x):.4f}, std={np.std(normalized_acc_x):.4f}")

    sax_params = params['SAX_PARAMS']

    segment_sax_strings = []
    segments_for_sax = []
    trimmed_acc_segs = []
    segment_metrics = []
    for idx, (s, e) in enumerate(acc_segs):
        seg_data = df[sax_channel + '_normalized'].iloc[s:e+1].to_numpy()
        seg_length = len(seg_data)
        trim_start = params['TRIM1']
        trim_end = seg_length - params['TRIM2']
        trimmed_length = seg_length - params['TRIM1'] - params['TRIM2']
        if trimmed_length < sax_params['WORD_LENGTH'] / 2:
            print(f"  Segment {idx} ({s}, {e}) too short after trimming (length={trimmed_length}, required={sax_params['WORD_LENGTH']/2}). Skipping SAX.")
            sax_str = "?" * sax_params['WORD_LENGTH']
        else:
            trimmed_seg_data = seg_data[trim_start:trim_end]
            print(f"Segment {idx} raw data (first 10 samples after trim): {trimmed_seg_data[:10]}")
            sax_str = SAXAlgorithm.series_to_sax(trimmed_seg_data, sax_params['WORD_LENGTH'], sax_params['ALPHABET_SIZE'], segment_idx=idx)
            segments_for_sax.append(trimmed_seg_data)
            trimmed_acc_segs.append((s + trim_start, s + trim_end - 1))
        segment_sax_strings.append(sax_str)
        stats = compute_stats(sax_str)
        segment_metrics.append({
            'Segment': f"Segment {idx+1}",
            'SAX': sax_str,
            'Changes': SAXAlgorithm.count_symbol_changes(sax_str),
            'Range': SAXAlgorithm.get_sax_symbol_range(sax_str),
            'Volatility': SAXAlgorithm.get_sax_value_volatility(sax_str),
            'Segments': 1,
            'Mean': stats['mean'],
            'Std': stats['std']
        })

    if not segments_for_sax:
        print(f"No valid trimmed segments for SAX analysis in {label}. Skipping SAX and plots.")
        return

    sax_data_dict[label] = segment_metrics

    segment_groups = []
    visited = [False] * len(segment_sax_strings)
    for i in range(len(segment_sax_strings)):
        csax = segment_sax_strings[i]
        if not visited[i] and "?" not in csax and "!" not in csax:
            grp_info = {'representative_sax': csax, 'segments_details': [{'sax_list_idx': i, 'original_indices': trimmed_acc_segs[i], 'actual_sax': csax}]}
            visited[i] = True
            for j in range(i + 1, len(segment_sax_strings)):
                osax = segment_sax_strings[j]
                if not visited[j] and "?" not in osax and "!" not in csax:
                    if SAXAlgorithm.levenshtein_distance(csax, osax) <= sax_params['SIMILARITY_THRESHOLD']:
                        grp_info['segments_details'].append({'sax_list_idx': j, 'original_indices': trimmed_acc_segs[j], 'actual_sax': osax})
                        visited[j] = True
            segment_groups.append(grp_info)

    print(f"Generated {len(segment_groups)} unique SAX pattern groups (before filtering).")

    filtered_segment_groups = []
    print(f"\n--- SAX Pattern Metrics & Filtering ({label}) ---")
    for grp_idx, group_info in enumerate(segment_groups):
        rep_sax = group_info['representative_sax']
        num_changes = SAXAlgorithm.count_symbol_changes(rep_sax)
        sym_range = SAXAlgorithm.get_sax_symbol_range(rep_sax)
        sym_volatility = SAXAlgorithm.get_sax_value_volatility(rep_sax)
        print(f"  Segment {grp_idx+1} (SAX: {rep_sax}): "
              f"Changes={num_changes}, Range={sym_range}, Volatility={sym_volatility}, "
              f"Segments={len(group_info['segments_details'])}")
        group_info['metrics'] = {'changes': num_changes, 'range': sym_range, 'volatility': sym_volatility}

        filter_metric = sax_params.get('SAX_FILTER_METRIC')
        filter_gt = sax_params.get('SAX_FILTER_THRESHOLD_GT')
        filter_lt = sax_params.get('SAX_FILTER_THRESHOLD_LT')
        if filter_metric and (filter_gt is not None or filter_lt is not None):
            metric_val = group_info['metrics'].get(filter_metric)
            if metric_val is None:
                filtered_segment_groups.append(group_info)
                continue
            remove_flag = False
            if filter_gt is not None and metric_val > filter_gt:
                remove_flag = True
            if filter_lt is not None and metric_val < filter_lt:
                remove_flag = True
            if not remove_flag:
                filtered_segment_groups.append(group_info)
        else:
            filtered_segment_groups.append(group_info)

    print(f"Number of groups after filtering: {len(filtered_segment_groups)}")

    transform_map = {}
    if 'Acc_X' in all_channels:
        transform_map.update({ch: NormalizeFunction.tanh1_np for ch in ['Acc_X', 'Acc_Y', 'Acc_Z']})
        transform_map.update({ch: NormalizeFunction.tanh2_np for ch in ['PPG_G1', 'PPG_G2', 'PPG_R1', 'PPG_R2', 'PPG_IR1', 'PPG_IR2', 'adc5', 'adc7']})
    else:
        transform_map.update({ch: NormalizeFunction.tanh3_np for ch in ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z', 'GYR_X(degree/sec)', 'GYR_Y', 'GYR_Z', 'Mag_X', 'Mag_Y', 'Mag_Z']})
    transform_map[sax_channel] = lambda x: x

    t = df.index / sampling_rate
    plt.figure(figsize=(12, 4))
    plt.plot(t, df[sax_channel], label=f'{sax_channel} Signal', alpha=0.7, color='black')
    for idx, (s, e) in enumerate(trimmed_acc_segs):
        seg_t = t[s:e+1]
        seg_data = df[sax_channel].iloc[s:e+1]
        plt.plot(seg_t, seg_data, '.', markersize=3, label=f'Segment {idx+1}' if idx < 10 else None)
    plt.title(f'{sax_channel} SAX Segments - {label}')
    plt.xlabel('Time (s)')
    plt.ylabel(f'{sax_channel} (m/s^2)')
    plt.legend(fontsize='small', ncol=2)
    plt.grid(True)
    plt.show()

# --- Main Execution ---
if __name__ == '__main__':
    params_right_ring = {
        'WINDOW': 2,
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 10,
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 100,
        'TRIM1': 50,
        'TRIM2': 12,
        'SAX_PARAMS': {
            'WORD_LENGTH': 50,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 10,
            'SAX_FILTER_THRESHOLD_LT': 5,
        }
    }

    params_left_ring = {
        'WINDOW': 2,
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 10,
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 100,
        'TRIM1': 50,
        'TRIM2': 12,
        'SAX_PARAMS': {
            'WORD_LENGTH': 50,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    params_right_watch_imu = {
        'WINDOW': 2,
        'STD_QUANTILE': 0.95,
        'MIN_DURATION': 40,
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 400,
        'TRIM1': 200,
        'TRIM2': 48,
        'SAX_PARAMS': {
            'WORD_LENGTH': 50,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 10,
            'SAX_FILTER_THRESHOLD_LT': 5,
        }
    }

    params_left_watch_imu = {
        'WINDOW': 2,
        'STD_QUANTILE': 0.95,
        'MIN_DURATION': 40,
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 400,
        'TRIM1': 200,
        'TRIM2': 48,
        'SAX_PARAMS': {
            'WORD_LENGTH': 50,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    params_right_watch_ppg = {
        'WINDOW': 2,
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 50,
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 150,
        'TRIM1': 50,
        'TRIM2': 12,
        'SAX_PARAMS': {
            'WORD_LENGTH': 50,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 10,
            'SAX_FILTER_THRESHOLD_LT': 5,
        }
    }

    params_left_watch_ppg = {
        'WINDOW': 2,
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 50,
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 150,
        'TRIM1': 50,
        'TRIM2': 12,
        'SAX_PARAMS': {
            'WORD_LENGTH': 50,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    sax_data_dict = {}

    df_right_ring, df_left_ring = None, None
    try:
        df_right_ring = pd.read_csv(right_ring_path, encoding='utf-8')
        print(f"Right ring data loaded: {df_right_ring.shape}")
    except FileNotFoundError:
        print(f"Error: Right ring file not found at {right_ring_path}")
    except Exception as e:
        print(f"Error loading right ring data: {e}")

    try:
        df_left_ring = pd.read_csv(left_ring_path, encoding='utf-8')
        print(f"Left ring data loaded: {df_left_ring.shape}")
    except FileNotFoundError:
        print(f"Error: Left ring file not found at {left_ring_path}")
    except Exception as e:
        print(f"Error loading left ring data: {e}")

    df_right_watch_imu, df_right_watch_ppg, df_left_watch_imu, df_left_watch_ppg = None, None, None, None
    try:
        df_right_watch_imu = pd.read_csv(right_watch_imu, encoding='utf-8')
        print(f"Right watch IMU data loaded: {df_right_watch_imu.shape}")
    except FileNotFoundError:
        print(f"Error: Right watch IMU file not found at {right_watch_imu}")
    except Exception as e:
        print(f"Error loading right watch IMU data: {e}")

    try:
        df_right_watch_ppg = pd.read_csv(right_watch_ppg, encoding='utf-8')
        print(f"Right watch PPG data loaded: {df_right_watch_ppg.shape}")
    except FileNotFoundError:
        print(f"Error: Right watch PPG file not found at {right_watch_ppg}")
    except Exception as e:
        print(f"Error loading right watch PPG data: {e}")

    try:
        df_left_watch_imu = pd.read_csv(left_watch_imu, encoding='utf-8')
        print(f"Left watch IMU data loaded: {df_left_watch_imu.shape}")
    except FileNotFoundError:
        print(f"Error: Left watch IMU file not found at {left_watch_imu}")
    except Exception as e:
        print(f"Error loading left watch IMU data: {e}")

    try:
        df_left_watch_ppg = pd.read_csv(left_watch_ppg, encoding='utf-8')
        print(f"Left watch PPG data loaded: {df_left_watch_ppg.shape}")
    except FileNotFoundError:
        print(f"Error: Left watch PPG file not found at {left_watch_ppg}")
    except Exception as e:
        print(f"Error loading left watch PPG data: {e}")

    if df_right_ring is not None:
        process_and_plot(df_right_ring, "Right Ring", params_right_ring, ['Acc_X', 'Acc_Y', 'Acc_Z'], right_ring_channels, sampling_rate=25.0, sax_data_dict=sax_data_dict)

    if df_left_ring is not None:
        process_and_plot(df_left_ring, "Left Ring", params_left_ring, ['Acc_X', 'Acc_Y', 'Acc_Z'], left_ring_channels, sampling_rate=25.0, sax_data_dict=sax_data_dict)

    if df_right_watch_imu is not None:
        process_and_plot(df_right_watch_imu, "Right Watch IMU", params_right_watch_imu, ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z'], right_watch_imu_channels, sampling_rate=100.0, sax_data_dict=sax_data_dict)

    if df_right_watch_ppg is not None:
        process_and_plot(df_right_watch_ppg, "Right Watch PPG", params_right_watch_ppg, ['Acc_X', 'Acc_Y', 'Acc_Z'], right_watch_ppg_channels, sampling_rate=25.0, sax_data_dict=sax_data_dict)

    if df_left_watch_imu is not None:
        process_and_plot(df_left_watch_imu, "Left Watch IMU", params_left_watch_imu, ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z'], left_watch_imu_channels, sampling_rate=100.0, sax_data_dict=sax_data_dict)

    if df_left_watch_ppg is not None:
        process_and_plot(df_left_watch_ppg, "Left Watch PPG", params_left_watch_ppg, ['Acc_X', 'Acc_Y', 'Acc_Z'], left_watch_ppg_channels, sampling_rate=25.0, sax_data_dict=sax_data_dict)

    print("\n===== Removing Non-Overlapping Segment 8 =====")
    filtered_sax_data = remove_non_overlapping_segment(sax_data_dict, segment_to_remove=8)

    print("\n===== Removing Segments with Std < 1 =====")
    filtered_sax_data = remove_low_std_segments(filtered_sax_data, std_threshold=1.0)

    outlier_results = identify_outlier_segments(filtered_sax_data)
    for dataset_name, segments in filtered_sax_data.items():
        print(f"\n--- Updated SAX Pattern Metrics with Statistics ({dataset_name}) ---")
        outlier = outlier_results.get(dataset_name, {})
        scores = outlier.get('Scores', [0] * len(segments))
        for idx, segment in enumerate(segments):
            outlier_score = scores[idx] if idx < len(scores) else 0
            print(f"  {segment['Segment']} (SAX: {segment['SAX']}): "
                  f"Changes={segment['Changes']}, Range={segment['Range']}, "
                  f"Volatility={segment['Volatility']}, Mean={segment['Mean']:.4f}, "
                  f"Std={segment['Std']:.4f}, Outlier Score={outlier_score:.4f}")

    print("\nAll processing finished.")
```
