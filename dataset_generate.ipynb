```
import math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
from scipy.stats import norm

# --- SAX Algorithm Functions ---
def z_normalize(series):
    if len(series) == 0: return np.array([])
    mean = np.mean(series); std = np.std(series)
    if std == 0: return np.zeros_like(series)
    return (series - mean) / std

def paa(series, num_segments):
    if num_segments <= 0: raise ValueError("num_segments must be positive")
    n = len(series)
    if n == 0: return np.full(num_segments, np.nan)
    segments = np.array_split(series, num_segments)
    return np.array([np.mean(seg) for seg in segments])

def get_breakpoints(alphabet_size):
    if not isinstance(alphabet_size, int) or alphabet_size < 1:
        raise ValueError("alphabet_size must be a positive integer")
    if alphabet_size <= 1: return np.array([])
    return norm.ppf(np.linspace(1/alphabet_size, 1 - 1/alphabet_size, alphabet_size - 1))

def series_to_sax(series, word_length, alphabet_size, segment_idx=None):
    if word_length <= 0 or alphabet_size <= 0:
        raise ValueError("word_length and alphabet_size must be positive")
    if alphabet_size > 26:
        raise ValueError("alphabet_size must not exceed 26 for lowercase letters")
    if not isinstance(series, np.ndarray): series_np = np.array(series, dtype=np.float64)
    else: series_np = series.astype(np.float64)
    if len(series_np) == 0: return "!" * word_length
    paa_values = paa(series_np, word_length)
    if np.isnan(paa_values).any():
        if np.all(np.isnan(paa_values)): return "?" * word_length
        paa_values = np.nan_to_num(paa_values, nan=0.0)
    breakpoints = get_breakpoints(alphabet_size)
    if segment_idx is not None:
        print(f"Segment {segment_idx} PAA values: {paa_values}")
        print(f"Breakpoints: {breakpoints}")
    alphabet = [chr(ord('a') + i) for i in range(alphabet_size)]
    sax_list = []
    for val in paa_values:
        symbol = alphabet[np.searchsorted(breakpoints, val)]
        sax_list.append(symbol)
    sax_str = "".join(sax_list)
    if segment_idx is not None:
        print(f"Segment {segment_idx} SAX string: {sax_str}")
    return sax_str

def levenshtein_distance(s1, s2):
    if len(s1) < len(s2): return levenshtein_distance(s2, s1)
    if len(s2) == 0: return len(s1)
    prev_row = range(len(s2) + 1)
    for i, c1 in enumerate(s1):
        curr_row = [i + 1]
        for j, c2 in enumerate(s2):
            subs = prev_row[j] + (c1 != c2)
            curr_row.append(min(prev_row[j+1]+1, curr_row[j]+1, subs))
        prev_row = curr_row
    return prev_row[-1]

def count_symbol_changes(sax_str):
    if not sax_str or len(sax_str) < 2: return 0
    changes = 0
    for i in range(len(sax_str) - 1):
        if sax_str[i] != sax_str[i+1]:
            changes += 1
    return changes

def get_sax_symbol_range(sax_str):
    if not sax_str: return 0
    valid_chars = [s for s in sax_str if s.islower()]
    if not valid_chars: return 0
    ordinal_values = [ord(s) - ord('a') for s in valid_chars]
    return max(ordinal_values) - min(ordinal_values)

def get_sax_value_volatility(sax_str):
    if not sax_str or len(sax_str) < 2: return 0
    valid_chars = [s for s in sax_str if s.islower()]
    if len(valid_chars) < 2: return 0
    ordinal_values = [ord(s) - ord('a') for s in valid_chars]
    volatility = sum(abs(ordinal_values[i+1] - ordinal_values[i]) for i in range(len(ordinal_values)-1))
    return volatility

# --- File Paths ---
right_ring_path = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/right_ring/RingSpO2_Id000001052_Dev516C_NoDev_1745891550572_P09_20250429_105230_HRIBI.csv')
left_ring_path = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/left_ring/RingSpO2_Id000001052_DevF952_NoDev_1745891555676_P09_20250429_105235_HRIBI.csv')
right_watch_imu = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/left_watch/fusion_20240416-120255.csv')
right_watch_ppg = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/left_watch/AFELogging_R945USQU1AXC2_6H0K_1.0.57_20240416-120255.csv')
left_watch_imu = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/right_watch/fusion_20240617-150340.csv')
left_watch_ppg = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/right_watch/AFELogging_R945USQU1AXC2_6HHY_1.0.57_20240617-150340.csv')

# --- Channel Definitions ---
right_ring_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'PPG_G1', 'PPG_G2', 'PPG_R1', 'PPG_R2', 'PPG', 'PPG_IR2']
left_ring_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'PPG_G1', 'PPG_G2', 'PPG_R1', 'PPG_R2', 'PPG', 'PPG_IR2']
right_watch_imu_channels = ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z', 'GYR_X(degree/sec)', 'GYR_Y', 'GYR_Z', 'Mag_X', 'Mag_Y', 'Mag_Z']
right_watch_ppg_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'adc5', 'adc7']
left_watch_imu_channels = ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z', 'GYR_X(degree/sec)', 'GYR_Y', 'GYR_Z', 'Mag_X', 'Mag_Y', 'Mag_Z']
left_watch_ppg_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'adc5', 'adc7']

# --- Reusable Functions ---
def extract_acc_segments(df, channels, window, std_q, min_dur, val_thresh, length_thresh):
    if not all(ch in df.columns for ch in channels):
        print(f"  Error: Not all required Acc channels ({channels}) found in DataFrame columns: {df.columns.tolist()}")
        return []
    try:
        print(f"  Acc data stats: {df[channels].describe()}")
        rs = df[channels].rolling(window=window, center=True).std().ffill().bfill().fillna(0)
        print(f"  Rolling std stats: {rs.describe()}")
        th = rs.quantile(std_q)
        print(f"  Std threshold (quantile {std_q}): {th}")
        calm_conditions = []
        for ch in channels:
            if pd.isna(th[ch]):
                print(f"    Warning: Could not compute threshold for channel '{ch}'. Skipping channel for calm mask.")
                calm_conditions.append(pd.Series([False] * len(df), index=df.index))
            else:
                calm_conditions.append(rs[ch] < th[ch])
        if not calm_conditions:
            print("    Error: Could not compute thresholds for any Acc channel.")
            return []
        calm_mask = np.logical_and.reduce(calm_conditions)
        print(f"  Calm mask: {calm_mask.sum()} true values out of {len(calm_mask)}")
        calm_mask = pd.Series(calm_mask, index=df.index)
        diff = calm_mask.astype(int).diff().fillna(0)
        starts = np.where(diff == 1)[0]
        ends = np.where(diff == -1)[0]
        if len(calm_mask) > 0:
            if calm_mask.iloc[0]:
                starts = np.insert(starts, 0, 0)
            if len(ends) < len(starts) and calm_mask.iloc[-1]:
                ends = np.append(ends, len(calm_mask) - 1)
        min_len = min(len(starts), len(ends))
        starts = starts[:min_len]
        ends = ends[:min_len]
        segs = []
        for s, e in zip(starts, ends):
            if s >= e: continue
            length = e - s + 1
            if length < min_dur: continue
            segment_data = df[channels].iloc[s : e + 1]
            max_abs = segment_data.fillna(0).abs().max().max()
            if max_abs <= val_thresh:
                print(f"  Segment ({s}, {e}) discarded: max abs value {max_abs:.2f} <= {val_thresh}")
                continue
            if length <= length_thresh:
                print(f"  Segment ({s}, {e}) discarded: length {length} <= {length_thresh}")
                continue
            segs.append((s, e))
        return segs
    except Exception as e:
        print(f"Error during segment extraction: {e}")
        return []

def tanh1_np(x_np):
    if not isinstance(x_np, np.ndarray): x_np = np.array(x_np, dtype=np.float64)
    x_scaled = x_np * 1e-5
    v_clip = np.clip(x_scaled, -20.0, 20.0)
    out = np.tanh(v_clip)
    if len(out) > 1:
        out[1:] *= 2
    return out * 10000

def tanh2_np(x_np):
    if not isinstance(x_np, np.ndarray): x_np = np.array(x_np, dtype=np.float64)
    x_scaled = x_np * 1e-5
    v_clip = np.clip(x_scaled, -20.0, 20.0)
    out = np.tanh(v_clip)
    return out * 100

def tanh3_np(x_np):
    if not isinstance(x_np, np.ndarray): x_np = np.array(x_np, dtype=np.float64)
    x_scaled = x_np * 1e-5
    v_clip = np.clip(x_scaled, -20.0, 20.0)
    out = np.tanh(v_clip)
    return out * 1000

# --- Main Processing Function with SAX ---
def process_and_plot(df, label, params, acc_channels, all_channels, sampling_rate):
    print(f"\n===== Processing: {label} (Sampling Rate: {sampling_rate}Hz) =====")
    if df is None or df.empty:
        print("Input DataFrame is empty or None. Skipping.")
        return

    print(f"Input data shape: {df.shape}")
    print(f"Using parameters: {params}")
    print(f"Using Acc channels: {acc_channels}")
    print(f"Using all channels: {all_channels}")

    # Extract acceleration-based segments
    # print(f"Extracting segments based on Acc channels for {label}...")
    acc_segs = extract_acc_segments(
        df, acc_channels,
        window=params['WINDOW'],
        std_q=params['STD_QUANTILE'],
        min_dur=params['MIN_DURATION'],
        val_thresh=params['VAL_THRESH'],
        length_thresh=params['LENGTH_THRESH']
    )
    # print(f"Extracted {len(acc_segs)} segments: {acc_segs}")

    if not acc_segs:
        print(f"No valid segments found for {label}. Skipping SAX and plots.")
        return

    # SAX Processing on trimmed Acc_X
    sax_channel = acc_channels[0]
    if sax_channel not in df.columns:
        print(f"  Error: SAX channel '{sax_channel}' not found. Skipping SAX analysis.")
        return

    # Normalize the entire SAX channel signal once
    normalized_acc_x = z_normalize(df[sax_channel].to_numpy())
    df[sax_channel + '_normalized'] = normalized_acc_x
    print(f"Normalized {sax_channel} stats: mean={np.mean(normalized_acc_x):.4f}, std={np.std(normalized_acc_x):.4f}")

    sax_params = params.get('SAX_PARAMS', {
        'WORD_LENGTH': 12,
        'ALPHABET_SIZE': 8,
        'SIMILARITY_THRESHOLD': 2,
        'SAX_FILTER_METRIC': 'volatility',
        'SAX_FILTER_THRESHOLD_GT': 15,
        'SAX_FILTER_THRESHOLD_LT': 7
    })

    # SAX Transformation with Trimming
    segment_sax_strings = []
    segments_for_sax = []
    trimmed_acc_segs = []
    for idx, (s, e) in enumerate(acc_segs):
        seg_data = df[sax_channel + '_normalized'].iloc[s:e+1].to_numpy()
        seg_length = len(seg_data)
        trim_start = params['TRIM1']
        trim_end = seg_length - params['TRIM2']
        trimmed_length = seg_length - params['TRIM1'] - params['TRIM2']
        if trimmed_length < sax_params['WORD_LENGTH'] / 2:
            print(f"  Segment {idx} ({s}, {e}) too short after trimming (length={trimmed_length}, required={sax_params['WORD_LENGTH']/2}). Skipping SAX.")
            sax_str = "?" * sax_params['WORD_LENGTH']
        else:
            trimmed_seg_data = seg_data[trim_start:trim_end]
            print(f"Segment {idx} raw data (first 10 samples after trim): {trimmed_seg_data[:10]}")
            sax_str = series_to_sax(trimmed_seg_data, sax_params['WORD_LENGTH'], sax_params['ALPHABET_SIZE'], segment_idx=idx)
            segments_for_sax.append(trimmed_seg_data)
            trimmed_acc_segs.append((s + trim_start, s + trim_end - 1))
        segment_sax_strings.append(sax_str)

    if not segments_for_sax:
        print(f"No valid trimmed segments for SAX analysis in {label}. Skipping SAX and plots.")
        return

    # SAX Grouping
    segment_groups = []
    visited = [False] * len(segment_sax_strings)
    for i in range(len(segment_sax_strings)):
        csax = segment_sax_strings[i]
        if not visited[i] and "?" not in csax and "!" not in csax:
            grp_info = {'representative_sax': csax, 'segments_details': [{'sax_list_idx': i, 'original_indices': trimmed_acc_segs[i], 'actual_sax': csax}]}
            visited[i] = True
            for j in range(i + 1, len(segment_sax_strings)):
                osax = segment_sax_strings[j]
                if not visited[j] and "?" not in osax and "!" not in csax:
                    if levenshtein_distance(csax, osax) <= sax_params['SIMILARITY_THRESHOLD']:
                        grp_info['segments_details'].append({'sax_list_idx': j, 'original_indices': trimmed_acc_segs[j], 'actual_sax': osax})
                        visited[j] = True
            segment_groups.append(grp_info)

    print(f"Generated {len(segment_groups)} unique SAX pattern groups (before filtering).")

    # SAX Metric Filtering
    filtered_segment_groups = []
    print(f"\n--- SAX Pattern Metrics & Filtering ({label}) ---")
    for grp_idx, group_info in enumerate(segment_groups):
        rep_sax = group_info['representative_sax']
        num_changes = count_symbol_changes(rep_sax)
        sym_range = get_sax_symbol_range(rep_sax)
        sym_volatility = get_sax_value_volatility(rep_sax)
        print(f"  Segment {grp_idx+1} (SAX: {rep_sax}): "
              f"Changes={num_changes}, Range={sym_range}, Volatility={sym_volatility}, "
              f"Segments={len(group_info['segments_details'])}")
        group_info['metrics'] = {'changes': num_changes, 'range': sym_range, 'volatility': sym_volatility}

        filter_metric = sax_params.get('SAX_FILTER_METRIC')
        filter_gt = sax_params.get('SAX_FILTER_THRESHOLD_GT')
        filter_lt = sax_params.get('SAX_FILTER_THRESHOLD_LT')
        if filter_metric and (filter_gt is not None or filter_lt is not None):
            metric_val = group_info['metrics'].get(filter_metric)
            if metric_val is None:
                filtered_segment_groups.append(group_info)
                continue
            remove_flag = False
            if filter_gt is not None and metric_val > filter_gt:
                remove_flag = True
            if filter_lt is not None and metric_val < filter_lt:
                remove_flag = True
            if not remove_flag:
                filtered_segment_groups.append(group_info)
        else:
            filtered_segment_groups.append(group_info)

    print(f"Number of groups after filtering: {len(filtered_segment_groups)}")

    # Normalization function mapping
    transform_map = {}
    if 'Acc_X' in all_channels:  # Ring or PPG channels
        transform_map.update({ch: tanh1_np for ch in ['Acc_X', 'Acc_Y', 'Acc_Z']})
        transform_map.update({ch: tanh2_np for ch in ['PPG_G1', 'PPG_G2', 'PPG_R1', 'PPG_R2', 'PPG_IR1', 'PPG_IR2', 'adc5', 'adc7']})
    else:  # IMU channels
        transform_map.update({ch: tanh3_np for ch in ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z', 'GYR_X(degree/sec)', 'GYR_Y', 'GYR_Z', 'Mag_X', 'Mag_Y', 'Mag_Z']})
    transform_map[sax_channel] = lambda x: x

    # --- Plot 1: SAX Channel with Trimmed Segment Overlay ---
    t = df.index / sampling_rate
    plt.figure(figsize=(12, 4))
    plt.plot(t, df[sax_channel], label=f'{sax_channel} Signal', alpha=0.7, color='black')
    for idx, (s, e) in enumerate(trimmed_acc_segs):
        seg_t = t[s:e+1]
        seg_data = df[sax_channel].iloc[s:e+1]
        plt.plot(seg_t, seg_data, '.', markersize=3, label=f'Segment {idx+1}' if idx < 10 else None)
    plt.title(f'{sax_channel} SAX Segments - {label}')
    plt.xlabel('Time (s)')
    plt.ylabel(f'{sax_channel} (m/s^2)')
    plt.legend(fontsize='small', ncol=2)
    plt.grid(True)
    plt.show()

    # # --- Plot 2: SAX Group Segments (Trimmed) ---
    # for grp_idx, group_info in enumerate(filtered_segment_groups):
    #     num_seg_in_grp = len(group_info['segments_details'])
    #     max_plots = 5
    #     if num_seg_in_grp == 0: continue
    #     rows_plot = min(num_seg_in_grp, max_plots)
    #     fig, axes = plt.subplots(rows_plot, 1, figsize=(12, 2.5*rows_plot), sharex=False)
    #     axes = np.atleast_1d(axes)
    #     fig.suptitle(f"{label} - SAX Group {grp_idx+1}: Rep SAX='{group_info['representative_sax']}' ({num_seg_in_grp} segs)")
    #     for p_idx, seg_det in enumerate(group_info['segments_details'][:rows_plot]):
    #         s_list_idx = seg_det['sax_list_idx']
    #         s_o, e_o = seg_det['original_indices']
    #         act_sax = seg_det['actual_sax']
    #         data = segments_for_sax[s_list_idx]
    #         seg_t = t[s_o:s_o+len(data)]
    #         ax = axes[p_idx]
    #         ax.plot(seg_t, data, label=f'Trimmed Seg {s_o}-{e_o}', color='black')
    #         ax.set_title(f"SAX: {act_sax}", fontsize=10)
    #         ax.legend(loc='upper right', fontsize='small')
    #         ax.grid(True)
    #     for k in range(rows_plot, len(axes)):
    #         axes[k].axis('off')
    #     plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    #     plt.show()

    # # --- Plot 3: Segment Channels (3xN) ---
    # print(f"Generating channel plots for {label}...")
    # num_plot_cols = 3
    # num_plot_rows = math.ceil(len(all_channels) / num_plot_cols)
    # for idx, (s, e) in enumerate(acc_segs, 1):
    #     raw_seg = df.iloc[s : e + 1]
    #     trimmed_length = len(raw_seg) - params['TRIM1'] - params['TRIM2']
    #     if trimmed_length <= 0:
    #         print(f"  Segment {idx} (len {len(raw_seg)}) too short after trimming (TRIM1={params['TRIM1']}, TRIM2={params['TRIM2']}). Skipping.")
    #         continue
    #     fig, axes = plt.subplots(num_plot_rows, num_plot_cols, figsize=(15, 4 * num_plot_rows), sharex=True, squeeze=False)
    #     axes = axes.flatten()
    #     plot_idx = 0
    #     for ch in all_channels:
    #         if ch in raw_seg.columns:
    #             norm_full = transform_map.get(ch, lambda x: x)(raw_seg[ch].to_numpy())
    #             trim_start = params['TRIM1']
    #             trim_end = len(norm_full) - params['TRIM2']
    #             norm_trim = norm_full[trim_start : trim_end]
    #             idxs = raw_seg.index[trim_start : trim_end] / sampling_rate
    #             ax = axes[plot_idx]
    #             ax.plot(idxs, norm_trim)
    #             ax.set_title(ch)
    #             ax.grid(True, linestyle='--', alpha=0.6)
    #             plot_idx += 1
    #         else:
    #             print(f"  Warning: Channel '{ch}' not found in segment {idx}. Skipping subplot.")
    #     for i in range(plot_idx, len(axes)):
    #         axes[i].axis('off')
    #     fig.suptitle(
    #         f'{label} - Segment {idx}: samples {s}–{e} (Trimmed Length={trimmed_length})',
    #         fontsize=16
    #     )
    #     plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    #     plt.show()

# --- Main Execution ---
if __name__ == '__main__':
    # Parameters for rings (25Hz)
    params_right_ring = {
        'WINDOW': 2,  # 0.08s
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 10,  # 0.4s
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 100,  # 4s
        'TRIM1': 25,  # 1s
        'TRIM2': 4,  # 0.16s
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 10,
            'SAX_FILTER_THRESHOLD_LT': 5,
        }
    }

    params_left_ring = {
        'WINDOW': 2,  # 0.08s
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 10,  # 0.4s
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 100,  # 4s
        'TRIM1': 25,  # 1s
        'TRIM2': 4,  # 0.16s
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    # Parameters for watch IMU (100Hz)
    params_right_watch_imu = {
        'WINDOW': 2,  # 0.08s
        'STD_QUANTILE': 0.95,
        'MIN_DURATION': 40,  # 0.4s
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 400,  # 4s
        'TRIM1': 100,  # 1s
        'TRIM2': 16,  # 0.16s
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 10,
            'SAX_FILTER_THRESHOLD_LT': 5,
        }
    }

    params_left_watch_imu = {
        'WINDOW': 2,  # 0.08s
        'STD_QUANTILE': 0.95,
        'MIN_DURATION': 40,  # 0.4s
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 400,  # 4s
        'TRIM1': 100,  # 1s
        'TRIM2': 16,  # 0.16s
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    # Parameters for watch PPG (25Hz)
    params_right_watch_ppg = {
        'WINDOW': 2,  # 0.08s
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 50,  # 0.4s
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 150,  # 4s
        'TRIM1': 25,  # 1s
        'TRIM2': 4,  # 0.16s
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 10,
            'SAX_FILTER_THRESHOLD_LT': 5,
        }
    }

    params_left_watch_ppg = {
        'WINDOW': 2,  # 0.08s
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 50,  # 0.4s
        'VAL_THRESH': 5,
        'LENGTH_THRESH': 150,  # 4s
        'TRIM1': 25,  # 1s
        'TRIM2': 4,  # 0.16s
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    # Load ring data (25Hz)
    df_right_ring, df_left_ring = None, None
    try:
        df_right_ring = pd.read_csv(right_ring_path, encoding='utf-8')
        print(f"Right ring data loaded: {df_right_ring.shape}")
    except FileNotFoundError:
        print(f"Error: Right ring file not found at {right_ring_path}")
    except Exception as e:
        print(f"Error loading right ring data: {e}")

    try:
        df_left_ring = pd.read_csv(left_ring_path, encoding='utf-8')
        print(f"Left ring data loaded: {df_left_ring.shape}")
    except FileNotFoundError:
        print(f"Error: Left ring file not found at {left_ring_path}")
    except Exception as e:
        print(f"Error loading left ring data: {e}")

    # Load watch data
    df_right_watch_imu, df_right_watch_ppg, df_left_watch_imu, df_left_watch_ppg = None, None, None, None
    try:
        df_right_watch_imu = pd.read_csv(right_watch_imu, encoding='utf-8')
        print(f"Right watch IMU data loaded: {df_right_watch_imu.shape}")
    except FileNotFoundError:
        print(f"Error: Right watch IMU file not found at {right_watch_imu}")
    except Exception as e:
        print(f"Error loading right watch IMU data: {e}")

    try:
        df_right_watch_ppg = pd.read_csv(right_watch_ppg, encoding='utf-8')
        print(f"Right watch PPG data loaded: {df_right_watch_ppg.shape}")
    except FileNotFoundError:
        print(f"Error: Right watch PPG file not found at {right_watch_ppg}")
    except Exception as e:
        print(f"Error loading right watch PPG data: {e}")

    try:
        df_left_watch_imu = pd.read_csv(left_watch_imu, encoding='utf-8')
        print(f"Left watch IMU data loaded: {df_left_watch_imu.shape}")
    except FileNotFoundError:
        print(f"Error: Left watch IMU file not found at {left_watch_imu}")
    except Exception as e:
        print(f"Error loading left watch IMU data: {e}")

    try:
        df_left_watch_ppg = pd.read_csv(left_watch_ppg, encoding='utf-8')
        print(f"Left watch PPG data loaded: {df_left_watch_ppg.shape}")
    except FileNotFoundError:
        print(f"Error: Left watch PPG file not found at {left_watch_ppg}")
    except Exception as e:
        print(f"Error loading left watch PPG data: {e}")

    # Process all data
    if df_right_ring is not None:
        process_and_plot(df_right_ring, "Right Ring", params_right_ring, ['Acc_X', 'Acc_Y', 'Acc_Z'], right_ring_channels, sampling_rate=25.0)

    if df_left_ring is not None:
        process_and_plot(df_left_ring, "Left Ring", params_left_ring, ['Acc_X', 'Acc_Y', 'Acc_Z'], left_ring_channels, sampling_rate=25.0)

    if df_right_watch_imu is not None:
        process_and_plot(df_right_watch_imu, "Right Watch IMU", params_right_watch_imu, ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z'], right_watch_imu_channels, sampling_rate=100.0)

    if df_right_watch_ppg is not None:
        process_and_plot(df_right_watch_ppg, "Right Watch PPG", params_right_watch_ppg, ['Acc_X', 'Acc_Y', 'Acc_Z'], right_watch_ppg_channels, sampling_rate=25.0)

    if df_left_watch_imu is not None:
        process_and_plot(df_left_watch_imu, "Left Watch IMU", params_left_watch_imu, ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z'], left_watch_imu_channels, sampling_rate=100.0)

    if df_left_watch_ppg is not None:
        process_and_plot(df_left_watch_ppg, "Left Watch PPG", params_left_watch_ppg, ['Acc_X', 'Acc_Y', 'Acc_Z'], left_watch_ppg_channels, sampling_rate=25.0)

    print("\nAll processing finished.")
```

```
import math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
from scipy.stats import norm

# --- SAX Algorithm Functions ---
def z_normalize(series):
    if len(series) == 0: return np.array([])
    mean = np.mean(series); std = np.std(series)
    if std == 0: return np.zeros_like(series)
    return (series - mean) / std

def paa(series, num_segments):
    if num_segments <= 0: raise ValueError("num_segments must be positive")
    n = len(series)
    if n == 0: return np.full(num_segments, np.nan)
    segments = np.array_split(series, num_segments)
    return np.array([np.mean(seg) for seg in segments])

def get_breakpoints(alphabet_size):
    if not isinstance(alphabet_size, int) or alphabet_size < 1:
        raise ValueError("alphabet_size must be a positive integer")
    if alphabet_size <= 1: return np.array([])
    return norm.ppf(np.linspace(1/alphabet_size, 1 - 1/alphabet_size, alphabet_size - 1))

def series_to_sax(series, word_length, alphabet_size, segment_idx=None):
    if word_length <= 0 or alphabet_size <= 0:
        raise ValueError("word_length and alphabet_size must be positive")
    if alphabet_size > 26:
        raise ValueError("alphabet_size must not exceed 26 for lowercase letters")
    if not isinstance(series, np.ndarray): series_np = np.array(series, dtype=np.float64)
    else: series_np = series.astype(np.float64)
    if len(series_np) == 0: return "!" * word_length
    paa_values = paa(series_np, word_length)
    if np.isnan(paa_values).any():
        if np.all(np.isnan(paa_values)): return "?" * word_length
        paa_values = np.nan_to_num(paa_values, nan=0.0)
    breakpoints = get_breakpoints(alphabet_size)
    if segment_idx is not None:
        print(f"Segment {segment_idx} PAA values: {paa_values}")
        print(f"Breakpoints: {breakpoints}")
    alphabet = [chr(ord('a') + i) for i in range(alphabet_size)]
    sax_list = []
    for val in paa_values:
        symbol = alphabet[np.searchsorted(breakpoints, val)]
        sax_list.append(symbol)
    sax_str = "".join(sax_list)
    if segment_idx is not None:
        print(f"Segment {segment_idx} SAX string: {sax_str}")
    return sax_str

def levenshtein_distance(s1, s2):
    if len(s1) < len(s2): return levenshtein_distance(s2, s1)
    if len(s2) == 0: return len(s1)
    prev_row = range(len(s2) + 1)
    for i, c1 in enumerate(s1):
        curr_row = [i + 1]
        for j, c2 in enumerate(s2):
            subs = prev_row[j] + (c1 != c2)
            curr_row.append(min(prev_row[j+1]+1, curr_row[j]+1, subs))
        prev_row = curr_row
    return prev_row[-1]

def count_symbol_changes(sax_str):
    if not sax_str or len(sax_str) < 2: return 0
    changes = 0
    for i in range(len(sax_str) - 1):
        if sax_str[i] != sax_str[i+1]:
            changes += 1
    return changes

def get_sax_symbol_range(sax_str):
    if not sax_str: return 0
    valid_chars = [s for s in sax_str if s.islower()]
    if not valid_chars: return 0
    ordinal_values = [ord(s) - ord('a') for s in valid_chars]
    return max(ordinal_values) - min(ordinal_values)

def get_sax_value_volatility(sax_str):
    if not sax_str or len(sax_str) < 2: return 0
    valid_chars = [s for s in sax_str if s.islower()]
    if len(valid_chars) < 2: return 0
    ordinal_values = [ord(s) - ord('a') for s in valid_chars]
    volatility = sum(abs(ordinal_values[i+1] - ordinal_values[i]) for i in range(len(ordinal_values)-1))
    return volatility

# --- File Paths ---
right_ring_path = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/right_ring/RingSpO2_Id000001052_Dev516C_NoDev_1745891550572_P09_20250429_105230_HRIBI.csv')
left_ring_path = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/left_ring/RingSpO2_Id000001052_DevF952_NoDev_1745891555676_P09_20250429_105235_HRIBI.csv')
right_watch_imu = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/left_watch/fusion_20240416-120255.csv')
right_watch_ppg = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/left_watch/AFELogging_R945USQU1AXC2_6H0K_1.0.57_20240416-120255.csv')
left_watch_imu = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/right_watch/fusion_20240617-150340.csv')
left_watch_ppg = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/right_watch/AFELogging_R945USQU1AXC2_6HHY_1.0.57_20240617-150340.csv')

# --- Channel Definitions ---
right_ring_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'PPG_G1', 'PPG_G2', 'PPG_R1', 'PPG_R2', 'PPG', 'PPG_IR2']
left_ring_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'PPG_G1', 'PPG_G2', 'PPG_R1', 'PPG_R2', 'PPG', 'PPG_IR2']
right_watch_imu_channels = ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z', 'GYR_X(degree/sec)', 'GYR_Y', 'GYR_Z', 'Mag_X', 'Mag_Y', 'Mag_Z']
left_watch_imu_channels = ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z', 'GYR_X(degree/sec)', 'GYR_Y', 'GYR_Z', 'Mag_X', 'Mag_Y', 'Mag_Z']
right_watch_ppg_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'adc5', 'adc7']
left_watch_ppg_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'adc5', 'adc7']

# --- Reusable Functions ---
def extract_acc_segments(df, channels, window, std_q, min_dur, val_thresh, length_thresh):
    if not all(ch in df.columns for ch in channels):
        print(f"  Error: Not all required Acc channels ({channels}) found in DataFrame columns: {df.columns.tolist()}")
        return []
    try:
        print(f"  Acc data stats: {df[channels].describe()}")
        rs = df[channels].rolling(window=window, center=True).std().ffill().bfill().fillna(0)
        print(f"  Rolling std stats: {rs.describe()}")
        th = rs.quantile(std_q)
        print(f"  Std threshold (quantile {std_q}): {th}")
        calm_conditions = []
        for ch in channels:
            if pd.isna(th[ch]):
                print(f"    Warning: Could not compute threshold for channel '{ch}'. Skipping channel for calm mask.")
                calm_conditions.append(pd.Series([False] * len(df), index=df.index))
            else:
                calm_conditions.append(rs[ch] < th[ch])
        if not calm_conditions:
            print("    Error: Could not compute thresholds for any Acc channel.")
            return []
        calm_mask = np.logical_and.reduce(calm_conditions)
        print(f"  Calm mask: {calm_mask.sum()} true values out of {len(calm_mask)}")
        calm_mask = pd.Series(calm_mask, index=df.index)
        diff = calm_mask.astype(int).diff().fillna(0)
        starts = np.where(diff == 1)[0]
        ends = np.where(diff == -1)[0]
        if len(calm_mask) > 0:
            if calm_mask.iloc[0]:
                starts = np.insert(starts, 0, 0)
            if len(ends) < len(starts) and calm_mask.iloc[-1]:
                ends = np.append(ends, len(calm_mask) - 1)
        min_len = min(len(starts), len(ends))
        starts = starts[:min_len]
        ends = ends[:min_len]
        segs = []
        for s, e in zip(starts, ends):
            if s >= e:
                # print(f"  Segment ({s}, {e}) discarded: invalid range")
                continue
            length = e - s + 1
            if length < min_dur:
                # print(f"  Segment ({s}, {e}) discarded: length {length} < {min_dur}")
                continue
            segment_data = df[channels].iloc[s : e + 1]
            max_abs = segment_data.fillna(0).abs().max().max()
            if max_abs <= val_thresh:
                # print(f"  Segment ({s}, {e}) discarded: max abs value {max_abs:.2f} <= {val_thresh}")
                continue
            if length <= length_thresh:
                # print(f"  Segment ({s}, {e}) discarded: length {length} <= {length_thresh}")
                continue
            segs.append((s, e))
        return segs
    except Exception as e:
        print(f"Error during segment extraction: {e}")
        return []

def tanh1_np(x_np):
    if not isinstance(x_np, np.ndarray): x_np = np.array(x_np, dtype=np.float64)
    x_scaled = x_np * 1e-5
    v_clip = np.clip(x_scaled, -20.0, 20.0)
    out = np.tanh(v_clip)
    if len(out) > 1:
        out[1:] *= 2
    return out * 10000

def tanh2_np(x_np):
    if not isinstance(x_np, np.ndarray): x_np = np.array(x_np, dtype=np.float64)
    x_scaled = x_np * 1e-5
    v_clip = np.clip(x_scaled, -20.0, 20.0)
    out = np.tanh(v_clip)
    return out * 100

def tanh3_np(x_np):
    if not isinstance(x_np, np.ndarray): x_np = np.array(x_np, dtype=np.float64)
    x_scaled = x_np * 1e-5
    v_clip = np.clip(x_scaled, -20.0, 20.0)
    out = np.tanh(v_clip)
    return out * 1000

# --- Main Processing Function with SAX ---
def process_and_plot(df, label, params, acc_channels, all_channels, sampling_rate):
    print(f"\n===== Processing: {label} (Sampling Rate: {sampling_rate}Hz) =====")
    if df is None or df.empty:
        print("Input DataFrame is empty or None. Skipping.")
        return []

    print(f"Input data shape: {df.shape}")
    print(f"Using parameters: {params}")
    print(f"Using Acc channels: {acc_channels}")
    print(f"Using all channels: {all_channels}")

    # Extract acceleration-based segments
    acc_segs = extract_acc_segments(
        df, acc_channels,
        window=params['WINDOW'],
        std_q=params['STD_QUANTILE'],
        min_dur=params['MIN_DURATION'],
        val_thresh=params['VAL_THRESH'],
        length_thresh=params['LENGTH_THRESH']
    )
    print(f"Extracted {len(acc_segs)} segments: {acc_segs}")

    if not acc_segs:
        print(f"No valid segments found for {label}. Skipping SAX and plots.")
        return []

    # SAX Processing on trimmed Acc_X
    sax_channel = acc_channels[0]
    if sax_channel not in df.columns:
        print(f"  Error: SAX channel '{sax_channel}' not found. Skipping SAX analysis.")
        return []

    # Normalize the entire SAX channel signal once
    normalized_acc_x = z_normalize(df[sax_channel].to_numpy())
    df[sax_channel + '_normalized'] = normalized_acc_x
    print(f"Normalized {sax_channel} stats: mean={np.mean(normalized_acc_x):.4f}, std={np.std(normalized_acc_x):.4f}")

    sax_params = params.get('SAX_PARAMS', {
        'WORD_LENGTH': 12,
        'ALPHABET_SIZE': 8,
        'SIMILARITY_THRESHOLD': 2,
        'SAX_FILTER_METRIC': 'volatility',
        'SAX_FILTER_THRESHOLD_GT': 15,
        'SAX_FILTER_THRESHOLD_LT': 7
    })

    # SAX Transformation with Trimming
    segment_sax_strings = []
    segments_for_sax = []
    trimmed_acc_segs = []
    segment_volatilities = []
    for idx, (s, e) in enumerate(acc_segs):
        seg_data = df[sax_channel + '_normalized'].iloc[s:e+1].to_numpy()
        seg_length = len(seg_data)
        trim_start = params['TRIM1']
        trim_end = seg_length - params['TRIM2']
        trimmed_length = seg_length - params['TRIM1'] - params['TRIM2']
        if trimmed_length < sax_params['WORD_LENGTH'] / 2:
            print(f"  Segment {idx} ({s}, {e}) too short after trimming (length={trimmed_length}, required={sax_params['WORD_LENGTH']/2}). Skipping SAX.")
            sax_str = "?" * sax_params['WORD_LENGTH']
            volatility = 0  # Placeholder for skipped segments
        else:
            trimmed_seg_data = seg_data[trim_start:trim_end]
            print(f"Segment {idx} raw data (first 10 samples after trim): {trimmed_seg_data[:10]}")
            sax_str = series_to_sax(trimmed_seg_data, sax_params['WORD_LENGTH'], sax_params['ALPHABET_SIZE'], segment_idx=idx)
            volatility = get_sax_value_volatility(sax_str)
            segments_for_sax.append(trimmed_seg_data)
            trimmed_acc_segs.append((s + trim_start, s + trim_end - 1))
        segment_sax_strings.append(sax_str)
        segment_volatilities.append(volatility)

    if not segments_for_sax:
        print(f"No valid trimmed segments for SAX analysis in {label}. Skipping SAX and plots.")
        return []

    # SAX Grouping
    segment_groups = []
    visited = [False] * len(segment_sax_strings)
    for i in range(len(segment_sax_strings)):
        csax = segment_sax_strings[i]
        if not visited[i] and "?" not in csax and "!" not in csax:
            grp_info = {'representative_sax': csax, 'segments_details': [{'sax_list_idx': i, 'original_indices': trimmed_acc_segs[i], 'actual_sax': csax}]}
            visited[i] = True
            for j in range(i + 1, len(segment_sax_strings)):
                osax = segment_sax_strings[j]
                if not visited[j] and "?" not in osax and "!" not in csax:
                    if levenshtein_distance(csax, osax) <= sax_params['SIMILARITY_THRESHOLD']:
                        grp_info['segments_details'].append({'sax_list_idx': j, 'original_indices': trimmed_acc_segs[j], 'actual_sax': osax})
                        visited[j] = True
            segment_groups.append(grp_info)

    print(f"Generated {len(segment_groups)} unique SAX pattern groups (before filtering).")

    # SAX Metric Filtering
    filtered_segment_groups = []
    print(f"\n--- SAX Pattern Metrics & Filtering ({label}) ---")
    for grp_idx, group_info in enumerate(segment_groups):
        rep_sax = group_info['representative_sax']
        num_changes = count_symbol_changes(rep_sax)
        sym_range = get_sax_symbol_range(rep_sax)
        sym_volatility = get_sax_value_volatility(rep_sax)
        print(f"  Segment {grp_idx+1} (SAX: {rep_sax}): "
              f"Changes={num_changes}, Range={sym_range}, Volatility={sym_volatility}, "
              f"Segments={len(group_info['segments_details'])}")
        group_info['metrics'] = {'changes': num_changes, 'range': sym_range, 'volatility': sym_volatility}

        filter_metric = sax_params.get('SAX_FILTER_METRIC')
        filter_gt = sax_params.get('SAX_FILTER_THRESHOLD_GT')
        filter_lt = sax_params.get('SAX_FILTER_THRESHOLD_LT')
        if filter_metric and (filter_gt is not None or filter_lt is not None):
            metric_val = group_info['metrics'].get(filter_metric)
            if metric_val is None:
                filtered_segment_groups.append(group_info)
                continue
            remove_flag = False
            if filter_gt is not None and metric_val > filter_gt:
                remove_flag = True
            if filter_lt is not None and metric_val < filter_lt:
                remove_flag = True
            if not remove_flag:
                filtered_segment_groups.append(group_info)
        else:
            filtered_segment_groups.append(group_info)

    print(f"Number of groups after filtering: {len(filtered_segment_groups)}")

    # Normalization function mapping
    transform_map = {}
    if 'Acc_X' in all_channels:  # Ring or PPG channels
        transform_map.update({ch: tanh1_np for ch in ['Acc_X', 'Acc_Y', 'Acc_Z']})
        transform_map.update({ch: tanh2_np for ch in ['PPG_G1', 'PPG_G2', 'PPG_R1', 'PPG_R2', 'PPG_IR1', 'PPG_IR2', 'adc5', 'adc7']})
    else:  # IMU channels
        transform_map.update({ch: tanh3_np for ch in ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z', 'GYR_X(degree/sec)', 'GYR_Y', 'GYR_Z', 'Mag_X', 'Mag_Y', 'Mag_Z']})
    transform_map[sax_channel] = lambda x: x

    # --- Plot 1: SAX Channel with Trimmed Segment Overlay ---
    t = df.index / sampling_rate
    plt.figure(figsize=(12, 4))
    plt.plot(t, df[sax_channel], label=f'{sax_channel} Signal', alpha=0.7, color='black')
    for idx, (s, e) in enumerate(trimmed_acc_segs):
        seg_t = t[s:e+1]
        seg_data = df[sax_channel].iloc[s:e+1]
        plt.plot(seg_t, seg_data, '.', markersize=3, label=f'Segment {idx+1}' if idx < 10 else None)
    plt.title(f'{sax_channel} SAX Segments - {label}')
    plt.xlabel('Time (s)')
    plt.ylabel(f'{sax_channel} (m/s^2)')
    plt.legend(fontsize='small', ncol=2)
    plt.grid(True)
    plt.show()

    return segment_volatilities

# --- Main Execution ---
if __name__ == '__main__':
    # Parameters for rings (25Hz)
    params_right_ring = {
        'WINDOW': 2,  # 0.08s
        'STD_QUANTILE': 0.85,
        'MIN_DURATION': 5,  # 0.2s
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 75,  # 3s
        'TRIM1': 10,  # 0.4s
        'TRIM2': 2,  # 0.08s
        'SAX_PARAMS': {
            'WORD_LENGTH': 20,  # Match provided metrics
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 10,
            'SAX_FILTER_THRESHOLD_LT': 5,
        }
    }

    params_left_ring = {
        'WINDOW': 2,  # 0.08s
        'STD_QUANTILE': 0.85,
        'MIN_DURATION': 5,  # 0.2s
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 75,  # 3s
        'TRIM1': 10,  # 0.4s
        'TRIM2': 2,  # 0.08s
        'SAX_PARAMS': {
            'WORD_LENGTH': 20,  # Match provided metrics
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    # Parameters for watch IMU (100Hz)
    params_right_watch_imu = {
        'WINDOW': 8,  # 0.08s
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 20,  # 0.2s
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 300,  # 3s
        'TRIM1': 40,  # 0.4s
        'TRIM2': 8,  # 0.08s
        'SAX_PARAMS': {
            'WORD_LENGTH': 20,  # Match provided metrics
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 10,
            'SAX_FILTER_THRESHOLD_LT': 5,
        }
    }

    params_left_watch_imu = {
        'WINDOW': 8,  # 0.08s
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 20,  # 0.2s
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 300,  # 3s
        'TRIM1': 40,  # 0.4s
        'TRIM2': 8,  # 0.08s
        'SAX_PARAMS': {
            'WORD_LENGTH': 20,  # Match provided metrics
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    # Parameters for watch PPG (25Hz)
    params_right_watch_ppg = {
        'WINDOW': 2,  # 0.08s
        'STD_QUANTILE': 0.85,
        'MIN_DURATION': 5,  # 0.2s
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 75,  # 3s
        'TRIM1': 10,  # 0.4s
        'TRIM2': 2,  # 0.08s
        'SAX_PARAMS': {
            'WORD_LENGTH': 20,  # Match provided metrics
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 10,
            'SAX_FILTER_THRESHOLD_LT': 5,
        }
    }

    params_left_watch_ppg = {
        'WINDOW': 2,  # 0.08s
        'STD_QUANTILE': 0.85,
        'MIN_DURATION': 5,  # 0.2s
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 75,  # 3s
        'TRIM1': 10,  # 0.4s
        'TRIM2': 2,  # 0.08s
        'SAX_PARAMS': {
            'WORD_LENGTH': 20,  # Match provided metrics
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 1,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    # Load ring data (25Hz)
    df_right_ring, df_left_ring = None, None
    try:
        df_right_ring = pd.read_csv(right_ring_path, encoding='utf-8')
        print(f"Right ring data loaded: {df_right_ring.shape}")
    except FileNotFoundError:
        print(f"Error: Right ring file not found at {right_ring_path}")
    except Exception as e:
        print(f"Error loading right ring data: {e}")

    try:
        df_left_ring = pd.read_csv(left_ring_path, encoding='utf-8')
        print(f"Left ring data loaded: {df_left_ring.shape}")
    except FileNotFoundError:
        print(f"Error: Left ring file not found at {left_ring_path}")
    except Exception as e:
        print(f"Error loading left ring data: {e}")

    # Load watch data
    df_right_watch_imu, df_right_watch_ppg, df_left_watch_imu, df_left_watch_ppg = None, None, None, None
    try:
        df_right_watch_imu = pd.read_csv(right_watch_imu, encoding='utf-8')
        print(f"Right watch IMU data loaded: {df_right_watch_imu.shape}")
    except FileNotFoundError:
        print(f"Error: Right watch IMU file not found at {right_watch_imu}")
    except Exception as e:
        print(f"Error loading right watch IMU data: {e}")

    try:
        df_right_watch_ppg = pd.read_csv(right_watch_ppg, encoding='utf-8')
        print(f"Right watch PPG data loaded: {df_right_watch_ppg.shape}")
    except FileNotFoundError:
        print(f"Error: Right watch PPG file not found at {right_watch_ppg}")
    except Exception as e:
        print(f"Error loading right watch PPG data: {e}")

    try:
        df_left_watch_imu = pd.read_csv(left_watch_imu, encoding='utf-8')
        print(f"Left watch IMU data loaded: {df_left_watch_imu.shape}")
    except FileNotFoundError:
        print(f"Error: Left watch IMU file not found at {left_watch_imu}")
    except Exception as e:
        print(f"Error loading left watch IMU data: {e}")

    try:
        df_left_watch_ppg = pd.read_csv(left_watch_ppg, encoding='utf-8')
        print(f"Left watch PPG data loaded: {df_left_watch_ppg.shape}")
    except FileNotFoundError:
        print(f"Error: Left watch PPG file not found at {left_watch_ppg}")
    except Exception as e:
        print(f"Error loading left watch PPG data: {e}")

    # Collect Volatility values from all datasets
    all_volatilities = []
    datasets = [
        ('Right Ring', df_right_ring, params_right_ring, ['Acc_X', 'Acc_Y', 'Acc_Z'], right_ring_channels, 25.0),
        ('Left Ring', df_left_ring, params_left_ring, ['Acc_X', 'Acc_Y', 'Acc_Z'], left_ring_channels, 25.0),
        ('Right Watch IMU', df_right_watch_imu, params_right_watch_imu, ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z'], right_watch_imu_channels, 100.0),
        ('Right Watch PPG', df_right_watch_ppg, params_right_watch_ppg, ['Acc_X', 'Acc_Y', 'Acc_Z'], right_watch_ppg_channels, 25.0),
        ('Left Watch IMU', df_left_watch_imu, params_left_watch_imu, ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z'], left_watch_imu_channels, 100.0),
        ('Left Watch PPG', df_left_watch_ppg, params_left_watch_ppg, ['Acc_X', 'Acc_Y', 'Acc_Z'], left_watch_ppg_channels, 25.0),
    ]

    for name, df, params, acc_ch, all_ch, sr in datasets:
        if df is not None:
            volatilities = process_and_plot(df, name, params, acc_ch, all_ch, sr)
            if volatilities:
                print(f"{name} has {len(volatilities)} segments")
            all_volatilities.append((name, volatilities))

    # Compute mean and standard deviation of Volatility for common segments
    print("\n--- Computing Mean and Standard Deviation of Volatility for Common Segments ---")
    segment_counts = [len(vols) for _, vols in all_volatilities if vols]
    if not segment_counts:
        print("No segments found in any dataset.")
    else:
        min_segments = min(segment_counts)
        if min_segments == 0:
            print("No common segments found.")
        else:
            for seg_idx in range(min_segments):
                volatilities = []
                dataset_names = []
                for name, vols in all_volatilities:
                    if len(vols) > seg_idx:
                        volatilities.append(vols[seg_idx])
                        dataset_names.append(name)
                if volatilities:
                    mean_vol = np.mean(volatilities)
                    std_vol = np.std(volatilities, ddof=1) if len(volatilities) > 1 else 0
                    print(f"Segment {seg_idx + 1}:")
                    print(f"  Datasets: {', '.join(dataset_names)}")
                    print(f"  Volatilities: {volatilities}")
                    print(f"  Mean Volatility: {mean_vol:.2f}")
                    print(f"  Standard Deviation of Volatility: {std_vol:.2f}\n")
                else:
                    print(f"Segment {seg_idx + 1}: No volatilities found.")

    print("\nAll processing finished.")
```

```
import math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
from scipy.stats import norm
from scipy.interpolate import interp1d

# --- SAX Algorithm Functions ---
def z_normalize(series):
    if len(series) == 0: return np.array([])
    mean = np.mean(series); std = np.std(series)
    if std == 0: return np.zeros_like(series)
    return (series - mean) / std

def paa(series, num_segments):
    if num_segments <= 0: raise ValueError("num_segments must be positive")
    n = len(series)
    if n == 0: return np.full(num_segments, np.nan)
    segments = np.array_split(series, num_segments)
    return np.array([np.mean(seg) for seg in segments])

def get_breakpoints(alphabet_size):
    if not isinstance(alphabet_size, int) or alphabet_size < 1:
        raise ValueError("alphabet_size must be a positive integer")
    if alphabet_size <= 1: return np.array([])
    return norm.ppf(np.linspace(1/alphabet_size, 1 - 1/alphabet_size, alphabet_size - 1))

def series_to_sax(series, word_length, alphabet_size, segment_idx=None):
    if word_length <= 0 or alphabet_size <= 0:
        raise ValueError("word_length and alphabet_size must be positive")
    if alphabet_size > 26:
        raise ValueError("alphabet_size must not exceed 26 for lowercase letters")
    if not isinstance(series, np.ndarray): series_np = np.array(series, dtype=np.float64)
    else: series_np = series.astype(np.float64)
    if len(series_np) == 0: return "!" * word_length
    paa_values = paa(series_np, word_length)
    if np.isnan(paa_values).any():
        if np.all(np.isnan(paa_values)): return "?" * word_length
        paa_values = np.nan_to_num(paa_values, nan=0.0)
    breakpoints = get_breakpoints(alphabet_size)
    if segment_idx is not None:
        print(f"Segment {segment_idx} PAA values: {paa_values}")
        print(f"Breakpoints: {breakpoints}")
    alphabet = [chr(ord('a') + i) for i in range(alphabet_size)]
    sax_list = []
    for val in paa_values:
        symbol = alphabet[np.searchsorted(breakpoints, val)]
        sax_list.append(symbol)
    sax_str = "".join(sax_list)
    if segment_idx is not None:
        print(f"Segment {segment_idx} SAX string: {sax_str}")
    return sax_str

def levenshtein_distance(s1, s2):
    if len(s1) < len(s2): return levenshtein_distance(s2, s1)
    if len(s2) == 0: return len(s1)
    prev_row = range(len(s2) + 1)
    for i, c1 in enumerate(s1):
        curr_row = [i + 1]
        for j, c2 in enumerate(s2):
            subs = prev_row[j] + (c1 != c2)
            curr_row.append(min(prev_row[j+1]+1, curr_row[j]+1, subs))
        prev_row = curr_row
    return prev_row[-1]

def count_symbol_changes(sax_str):
    if not sax_str or len(sax_str) < 2: return 0
    changes = 0
    for i in range(len(sax_str) - 1):
        if sax_str[i] != sax_str[i+1]:
            changes += 1
    return changes

def get_sax_symbol_range(sax_str):
    if not sax_str: return 0
    valid_chars = [s for s in sax_str if s.islower()]
    if not valid_chars: return 0
    ordinal_values = [ord(s) - ord('a') for s in valid_chars]
    return max(ordinal_values) - min(ordinal_values)

def get_sax_value_volatility(sax_str):
    if not sax_str or len(sax_str) < 2: return 0
    valid_chars = [s for s in sax_str if s.islower()]
    if len(valid_chars) < 2: return 0
    ordinal_values = [ord(s) - ord('a') for s in valid_chars]
    volatility = sum(abs(ordinal_values[i+1] - ordinal_values[i]) for i in range(len(ordinal_values)-1))
    return volatility

# --- File Paths ---
right_ring_path = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/right_ring/RingSpO2_Id000001052_Dev516C_NoDev_1745891550572_P09_20250529_173529_HRIBI.csv')
left_ring_path = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/left_ring/RingSpO2_Id000001052_DevF952_NoDev_1745891555676_P09_20250529_173529_HRIBI.csv')
right_watch_imu = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/left_watch/fusion_20240416-120255.csv')
right_watch_ppg = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/left_watch/AFELogging_R945USQU1AXC2_6H0K_1.0.57_20240416-120255.csv')
left_watch_imu = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/right_watch/fusion_20240617-150340.csv')
left_watch_ppg = os.path.join('/content/drive/MyDrive/watch_ring/250429/04April_2011_Monday_heute-956/right_watch/AFELogging_R945USQU1AXC2_6HHY_1.0.57_20240617-150340.csv')

# --- Channel Definitions ---
right_ring_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'PPG_G1', 'PPG_G2', 'PPG_R1', 'PPG_R2', 'PPG', 'PPG_IR2']
left_ring_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'PPG_G1', 'PPG_G2', 'PPG_R1', 'PPG_R2', 'PPG', 'PPG_IR2']
right_watch_imu_channels = ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z', 'GYR_X(degree/sec)', 'GYR_Y', 'GYR_Z', 'Mag_X', 'Mag_Y', 'Mag_Z']
left_watch_imu_channels = ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z', 'GYR_X(degree/sec)', 'GYR_Y', 'GYR_Z', 'Mag_X', 'Mag_Y', 'Mag_Z']
right_watch_ppg_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'adc5', 'adc7']
left_watch_ppg_channels = ['Acc_X', 'Acc_Y', 'Acc_Z', 'adc5', 'adc7']

# --- Reusable Functions ---
def extract_acc_segments(df, channels, window, std_q, min_dur, val_thresh, length_thresh):
    if not all(ch in df.columns for ch in channels):
        print(f"  Error: Not all required Acc channels ({channels}) found in DataFrame columns: {df.columns.tolist()}")
        return []
    try:
        print(f"  Acc data stats: {df[channels].describe()}")
        rs = df[channels].rolling(window=window, center=True).std().ffill().bfill().fillna(0)
        print(f"  Rolling std stats: {rs.describe()}")
        th = rs.quantile(std_q)
        print(f"  Std threshold (quantile {std_q}): {th}")
        calm_conditions = []
        for ch in channels:
            if pd.isna(th[ch]):
                print(f"    Warning: Could not compute threshold for channel '{ch}'. Skipping channel for calm mask.")
                calm_conditions.append(pd.Series([False] * len(df), index=df.index))
            else:
                calm_conditions.append(rs[ch] < th[ch])
        if not calm_conditions:
            print("    Error: Could not compute thresholds for any Acc channel.")
            return []
        calm_mask = np.logical_and.reduce(calm_conditions)
        print(f"  Calm mask: {calm_mask.sum()} true values out of {len(calm_mask)}")
        calm_mask = pd.Series(calm_mask, index=df.index)
        diff = calm_mask.astype(int).diff().fillna(0)
        starts = np.where(diff == 1)[0]
        ends = np.where(diff == -1)[0]
        if len(calm_mask) > 0:
            if calm_mask.iloc[0]:
                starts = np.insert(starts, 0, 0)
            if len(ends) < len(starts) and calm_mask.iloc[-1]:
                ends = np.append(ends, len(calm_mask) - 1)
        min_len = min(len(starts), len(ends))
        starts = starts[:min_len]
        ends = ends[:min_len]
        segs = []
        for s, e in zip(starts, ends):
            if s >= e:
                print(f"  Segment ({s}, {e}) discarded: invalid range")
                continue
            length = e - s + 1
            if length < min_dur:
                print(f"  Segment ({s}, {e}) discarded: length {length} < {min_dur}")
                continue
            segment_data = df[channels].iloc[s : e + 1]
            max_abs = segment_data.fillna(0).abs().max().max()
            if max_abs <= val_thresh:
                print(f"  Segment ({s}, {e}) discarded: max abs value {max_abs:.2f} <= {val_thresh}")
                continue
            if length <= length_thresh:
                print(f"  Segment ({s}, {e}) discarded: length {length} <= {length_thresh}")
                continue
            segs.append((s, e))
        return segs
    except Exception as e:
        print(f"Error during segment extraction: {e}")
        return []

def tanh1_np(x_np):
    if not isinstance(x_np, np.ndarray): x_np = np.array(x_np, dtype=np.float64)
    x_scaled = x_np * 1e-5
    v_clip = np.clip(x_scaled, -20.0, 20.0)
    out = np.tanh(v_clip)
    if len(out) > 1:
        out[1:] *= 2
    return out * 10000

def tanh2_np(x_np):
    if not isinstance(x_np, np.ndarray): x_np = np.array(x_np, dtype=np.float64)
    x_scaled = x_np * 1e-5
    v_clip = np.clip(x_scaled, -20.0, 20.0)
    out = np.tanh(v_clip)
    return out * 100

def tanh3_np(x_np):
    if not isinstance(x_np, np.ndarray): x_np = np.array(x_np, dtype=np.float64)
    x_scaled = x_np * 1e-5
    v_clip = np.clip(x_scaled, -20.0, 20.0)
    out = np.tanh(v_clip)
    return out * 1000

# --- Main Processing Function with SAX ---
def process_and_plot(df, label, params, acc_channels, all_channels, sampling_rate):
    print(f"\n===== Processing: {label} (Sampling Rate: {sampling_rate}Hz) =====")
    if df is None or df.empty:
        print("Input DataFrame is empty or None. Skipping.")
        return [], [], []

    print(f"Input data shape: {df.shape}")
    print(f"Using parameters: {params}")
    print(f"Using Acc channels: {acc_channels}")
    print(f"Using all channels: {all_channels}")

    # Extract acceleration-based segments
    acc_segs = extract_acc_segments(
        df, acc_channels,
        window=params['WINDOW'],
        std_q=params['STD_QUANTILE'],
        min_dur=params['MIN_DURATION'],
        val_thresh=params['VAL_THRESH'],
        length_thresh=params['LENGTH_THRESH']
    )
    print(f"Extracted {len(acc_segs)} segments: {acc_segs}")

    if not acc_segs:
        print(f"No valid segments found for {label}. Skipping SAX and plots.")
        return [], [], []

    # SAX Processing on trimmed Acc_X
    sax_channel = acc_channels[0]
    if sax_channel not in df.columns:
        print(f"  Error: SAX channel '{sax_channel}' not found. Skipping SAX analysis.")
        return [], [], []

    # Normalize the entire SAX channel signal once
    normalized_acc_x = z_normalize(df[sax_channel].to_numpy())
    df[sax_channel + '_normalized'] = normalized_acc_x
    print(f"Normalized {sax_channel} stats: mean={np.mean(normalized_acc_x):.4f}, std={np.std(normalized_acc_x):.4f}")

    sax_params = params.get('SAX_PARAMS', {
        'WORD_LENGTH': 10,
        'ALPHABET_SIZE': 8,
        'SIMILARITY_THRESHOLD': 2,
        'SAX_FILTER_METRIC': 'volatility',
        'SAX_FILTER_THRESHOLD_GT': 15,
        'SAX_FILTER_THRESHOLD_LT': 7
    })

    # SAX Transformation with Trimming
    segment_sax_strings = []
    segment_volatilities = []
    segment_data = []  # Store normalized segment data for plotting
    for idx, (s, e) in enumerate(acc_segs):
        seg_data = df[sax_channel + '_normalized'].iloc[s:e+1].to_numpy()
        seg_length = len(seg_data)
        trim_start = params['TRIM1']
        trim_end = seg_length - params['TRIM2']
        trimmed_length = seg_length - params['TRIM1'] - params['TRIM2']
        if trimmed_length < sax_params['WORD_LENGTH'] / 2:
            print(f"  Segment {idx} ({s}, {e}) too short after trimming (length={trimmed_length}, required={sax_params['WORD_LENGTH']/2}). Skipping SAX.")
            sax_str = "?" * sax_params['WORD_LENGTH']
            volatility = 0
            trimmed_seg_data = []
        else:
            trimmed_seg_data = seg_data[trim_start:trim_end]
            print(f"Segment {idx} raw data (first 10 samples after trim): {trimmed_seg_data[:10]}")
            sax_str = series_to_sax(trimmed_seg_data, sax_params['WORD_LENGTH'], sax_params['ALPHABET_SIZE'], segment_idx=idx)
            volatility = get_sax_value_volatility(sax_str)
        segment_sax_strings.append(sax_str)
        segment_volatilities.append(volatility)
        segment_data.append(trimmed_seg_data)

    if not any(segment_sax_strings):
        print(f"No valid trimmed segments for SAX analysis in {label}. Skipping.")
        return [], [], []

    # SAX Grouping (within dataset)
    segment_groups = []
    visited = [False] * len(segment_sax_strings)
    for i in range(len(segment_sax_strings)):
        csax = segment_sax_strings[i]
        if not visited[i] and "?" not in csax and "!" not in csax:
            grp_info = {'representative_sax': csax, 'segments_details': [{'sax_list_idx': i, 'actual_sax': csax}]}
            visited[i] = True
            for j in range(i + 1, len(segment_sax_strings)):
                osax = segment_sax_strings[j]
                if not visited[j] and "?" not in osax and "!" not in csax:
                    if levenshtein_distance(csax, osax) <= sax_params['SIMILARITY_THRESHOLD']:
                        grp_info['segments_details'].append({'sax_list_idx': j, 'actual_sax': osax})
                        visited[j] = True
            segment_groups.append(grp_info)

    print(f"Generated {len(segment_groups)} unique SAX pattern groups (before filtering).")

    # SAX Metric Filtering
    filtered_segment_groups = []
    print(f"\n--- SAX Pattern Metrics & Filtering ({label}) ---")
    for grp_idx, group_info in enumerate(segment_groups):
        rep_sax = group_info['representative_sax']
        num_changes = count_symbol_changes(rep_sax)
        sym_range = get_sax_symbol_range(rep_sax)
        sym_volatility = get_sax_value_volatility(rep_sax)
        print(f"  Segment {grp_idx+1} (SAX: {rep_sax}): "
              f"Changes={num_changes}, Range={sym_range}, Volatility={sym_volatility}, "
              f"Segments={len(group_info['segments_details'])}")
        group_info['metrics'] = {'changes': num_changes, 'range': sym_range, 'volatility': sym_volatility}

        filter_metric = sax_params.get('SAX_FILTER_METRIC')
        filter_gt = sax_params.get('SAX_FILTER_THRESHOLD_GT')
        filter_lt = sax_params.get('SAX_FILTER_THRESHOLD_LT')
        if filter_metric and (filter_gt is not None or filter_lt is not None):
            metric_val = group_info['metrics'].get(filter_metric)
            if metric_val is None:
                filtered_segment_groups.append(group_info)
                continue
            remove_flag = False
            if filter_gt is not None and metric_val > filter_gt:
                remove_flag = True
            if filter_lt is not None and metric_val < filter_lt:
                remove_flag = True
            if not remove_flag:
                filtered_segment_groups.append(group_info)
        else:
            filtered_segment_groups.append(group_info)

    print(f"Number of groups after filtering: {len(filtered_segment_groups)}")

    return segment_volatilities, segment_sax_strings, segment_data

# --- Main Execution ---
if __name__ == '__main__':
    # Parameters for rings (25Hz)
    params_right_ring = {
        'WINDOW': 2,
        'STD_QUANTILE': 0.85,
        'MIN_DURATION': 5,
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 75,
        'TRIM1': 10,
        'TRIM2': 2,
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 15,
            'SAX_FILTER_THRESHOLD_LT': 7,
        }
    }

    params_left_ring = {
        'WINDOW': 2,
        'STD_QUANTILE': 0.85,
        'MIN_DURATION': 5,
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 75,
        'TRIM1': 10,
        'TRIM2': 2,
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    # Parameters for watch IMU (100Hz)
    params_right_watch_imu = {
        'WINDOW': 8,
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 20,
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 300,
        'TRIM1': 40,
        'TRIM2': 8,
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 10,
            'SAX_FILTER_THRESHOLD_LT': 5,
        }
    }

    params_left_watch_imu = {
        'WINDOW': 8,
        'STD_QUANTILE': 0.9,
        'MIN_DURATION': 20,
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 300,
        'TRIM1': 40,
        'TRIM2': 8,
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    # Parameters for watch PPG (25Hz)
    params_right_watch_ppg = {
        'WINDOW': 2,
        'STD_QUANTILE': 0.85,
        'MIN_DURATION': 5,
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 75,
        'TRIM1': 10,
        'TRIM2': 2,
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'volatility',
            'SAX_FILTER_THRESHOLD_GT': 10,
            'SAX_FILTER_THRESHOLD_LT': 5,
        }
    }

    params_left_watch_ppg = {
        'WINDOW': 2,
        'STD_QUANTILE': 0.85,
        'MIN_DURATION': 5,
        'VAL_THRESH': 3,
        'LENGTH_THRESH': 75,
        'TRIM1': 10,
        'TRIM2': 2,
        'SAX_PARAMS': {
            'WORD_LENGTH': 10,
            'ALPHABET_SIZE': 8,
            'SIMILARITY_THRESHOLD': 2,
            'SAX_FILTER_METRIC': 'changes',
            'SAX_FILTER_THRESHOLD_GT': 5,
            'SAX_FILTER_THRESHOLD_LT': None
        }
    }

    # Load ring data (25Hz)
    df_right_ring, df_left_ring = None, None
    try:
        df_right_ring = pd.read_csv(right_ring_path, encoding='utf-8')
        print(f"Right ring data loaded: {df_right_ring.shape}")
    except FileNotFoundError:
        print(f"Error: Right ring file not found at {right_ring_path}")
    except Exception as e:
        print(f"Error loading right ring data: {e}")

    try:
        df_left_ring = pd.read_csv(left_ring_path, encoding='utf-8')
        print(f"Left ring data loaded: {df_left_ring.shape}")
    except FileNotFoundError:
        print(f"Error: Left ring file not found at {left_ring_path}")
    except Exception as e:
        print(f"Error loading left ring data: {e}")

    # Load watch data
    df_right_watch_imu, df_right_watch_ppg, df_left_watch_imu, df_left_watch_ppg = None, None, None, None
    try:
        df_right_watch_imu = pd.read_csv(right_watch_imu, encoding='utf-8')
        print(f"Right watch IMU data loaded: {df_right_watch_imu.shape}")
    except FileNotFoundError:
        print(f"Error: Right watch IMU file not found at {right_watch_imu}")
    except Exception as e:
        print(f"Error loading right watch IMU data: {e}")

    try:
        df_right_watch_ppg = pd.read_csv(right_watch_ppg, encoding='utf-8')
        print(f"Right watch PPG data loaded: {df_right_watch_ppg.shape}")
    except FileNotFoundError:
        print(f"Error: Right watch PPG file not found at {right_watch_ppg}")
    except Exception as e:
        print(f"Error loading right watch PPG data: {e}")

    try:
        df_left_watch_imu = pd.read_csv(left_watch_imu, encoding='utf-8')
        print(f"Left watch IMU data loaded: {df_left_watch_imu.shape}")
    except FileNotFoundError:
        print(f"Error: Left watch IMU file not found at {left_watch_imu}")
    except Exception as e:
        print(f"Error loading left watch IMU data: {e}")

    try:
        df_left_watch_ppg = pd.read_csv(left_watch_ppg, encoding='utf-8')
        print(f"Left watch PPG data loaded: {df_left_watch_ppg.shape}")
    except FileNotFoundError:
        print(f"Error: Left watch PPG file not found at {left_watch_ppg}")
    except Exception as e:
        print(f"Error loading left watch PPG data: {e}")

    # Collect Volatility values, SAX strings, and segment data from all datasets
    all_data = []
    datasets = [
        ('Right Ring', df_right_ring, params_right_ring, ['Acc_X', 'Acc_Y', 'Acc_Z'], right_ring_channels, 25.0),
        ('Left Ring', df_left_ring, params_left_ring, ['Acc_X', 'Acc_Y', 'Acc_Z'], left_ring_channels, 25.0),
        ('Right Watch IMU', df_right_watch_imu, params_right_watch_imu, ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z'], right_watch_imu_channels, 100.0),
        ('Right Watch PPG', df_right_watch_ppg, params_right_watch_ppg, ['Acc_X', 'Acc_Y', 'Acc_Z'], right_watch_ppg_channels, 25.0),
        ('Left Watch IMU', df_left_watch_imu, params_left_watch_imu, ['ACC_X(m/s^2)', 'ACC_Y', 'ACC_Z'], left_watch_imu_channels, 100.0),
        ('Left Watch PPG', df_left_watch_ppg, params_left_watch_ppg, ['Acc_X', 'Acc_Y', 'Acc_Z'], left_watch_ppg_channels, 25.0),
    ]

    for name, df, params, acc_ch, all_ch, sr in datasets:
        if df is not None:
            volatilities, sax_strings, seg_data = process_and_plot(df, name, params, acc_ch, all_ch, sr)
            if volatilities:
                print(f"{name} has {len(volatilities)} segments")
            all_data.append((name, volatilities, sax_strings, seg_data, sr, params))

    # Compute mean and standard deviation of all volatilities for common segments
    print("\n--- Computing Statistics for All Volatilities in Common Segments ---")
    segment_counts = [len(vols) for _, vols, _, _, _, _ in all_data if vols]
    if not segment_counts:
        print("No segments found in any dataset.")
    else:
        min_segments = min(segment_counts)
        if min_segments == 0:
            print("No common segments found.")
        else:
            # Collect all volatilities across common segments
            all_vols = []
            segment_volatilities = [[] for _ in range(min_segments)]
            for seg_idx in range(min_segments):
                for _, vols, _, _, _, _ in all_data:
                    if len(vols) > seg_idx:
                        all_vols.append(vols[seg_idx])
                        segment_volatilities[seg_idx].append(vols[seg_idx])

            if all_vols:
                mean_vol = np.mean(all_vols)
                std_vol = np.std(all_vols, ddof=1)

                # Set confidence level (90%)
                confidence_level = 0.90
                alpha = 1 - confidence_level
                z_score = norm.ppf(1 - alpha / 2)  # Z-score for 90% CI (≈1.645)

                # Compute confidence interval bounds
                lower_bound = mean_vol - z_score * std_vol
                upper_bound = mean_vol + z_score * std_vol

                # Identify segments with extreme volatilities
                segments_to_exclude = []
                for seg_idx, vols in enumerate(segment_volatilities):
                    if any(v < lower_bound or v > upper_bound for v in vols):
                        segments_to_exclude.append(seg_idx)

                print(f"Mean Volatility: {mean_vol:.2f}")
                print(f"Standard Deviation of Volatility: {std_vol:.2f}")
                print(f"{confidence_level*100:.0f}% Confidence Interval: ({lower_bound:.2f}, {upper_bound:.2f})")
                print(f"Segments with Extreme Volatilities (to exclude): {[i+1 for i in segments_to_exclude]}")
            else:
                print("No volatility values found.")
                segments_to_exclude = []

            # Process common segments, excluding those with extreme volatilities
            valid_segments = [i for i in range(min_segments) if i not in segments_to_exclude]
            if not valid_segments:
                print("No valid segments remain after excluding extreme volatilities.")
            else:
                print("\n--- Grouping and Plotting Similar Segments ---")
                # Group similar segments across datasets
                similar_groups = []
                common_length = 100  # Number of points for interpolation
                for seg_idx in valid_segments:
                    # Collect SAX strings and data for this segment across datasets
                    segment_info = []
                    for name, _, sax_strings, seg_data, sr, params in all_data:
                        if len(sax_strings) > seg_idx and "?" not in sax_strings[seg_idx] and "!" not in sax_strings[seg_idx]:
                            data = seg_data[seg_idx]
                            if len(data) > 0:
                                # Interpolate data to common length
                                x_old = np.linspace(0, 1, len(data))
                                x_new = np.linspace(0, 1, common_length)
                                interpolator = interp1d(x_old, data, kind='linear', fill_value="extrapolate")
                                interpolated_data = interpolator(x_new)
                                segment_info.append({
                                    'dataset': name,
                                    'sax': sax_strings[seg_idx],
                                    'data': interpolated_data,
                                    'sampling_rate': sr,
                                    'threshold': params['SAX_PARAMS']['SIMILARITY_THRESHOLD']
                                })

                    if len(segment_info) < 2:
                        print(f"Segment {seg_idx+1}: Not enough valid SAX strings for grouping ({len(segment_info)} datasets).")
                        continue

                    # Debug: Print SAX strings
                    print(f"Segment {seg_idx+1} SAX strings:")
                    for info in segment_info:
                        print(f"  {info['dataset']}: {info['sax']} (threshold={info['threshold']})")

                    # Group similar SAX strings
                    groups = []
                    visited = [False] * len(segment_info)
                    for i in range(len(segment_info)):
                        if not visited[i]:
                            group = [segment_info[i]]
                            visited[i] = True
                            for j in range(i + 1, len(segment_info)):
                                if not visited[j]:
                                    threshold = min(segment_info[i]['threshold'], segment_info[j]['threshold'])
                                    distance = levenshtein_distance(segment_info[i]['sax'], segment_info[j]['sax'])
                                    print(f"Comparing {segment_info[i]['dataset']} vs {segment_info[j]['dataset']}: "
                                          f"Distance={distance}, Threshold={threshold}")
                                    if distance <= threshold:
                                        group.append(segment_info[j])
                                        visited[j] = True
                            if len(group) > 1:  # Only include groups with multiple datasets
                                groups.append(group)

                    if groups:
                        similar_groups.append((seg_idx + 1, groups))
                    else:
                        print(f"Segment {seg_idx+1}: No similar segment groups found.")

                # Plot similar segments
                if not similar_groups:
                    print("No similar segment groups found across all segments.")
                else:
                    for seg_num, groups in similar_groups:
                        for group_idx, group in enumerate(groups, 1):
                            print(f"Plotting Segment {seg_num} - Group {group_idx} with {len(group)} datasets")
                            plt.figure(figsize=(12, 6))
                            t = np.linspace(0, 1, common_length)
                            for item in group:
                                data = item['data']
                                plt.plot(t, data, label=item['dataset'], linewidth=1)
                            plt.title(f"Similar Segments in Segment {seg_num} - Group {group_idx}")
                            plt.xlabel('Normalized Time')
                            plt.ylabel('Normalized Acc_X')
                            plt.legend()
                            plt.grid(True)
                            plt.tight_layout()
                            plt.show()

print("\nAll processing finished.")
```
